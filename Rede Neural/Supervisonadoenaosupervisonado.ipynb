```python
# realizando a construção dos gráficos

groups = df_supervisionado.groupby('Classe')

fig, ax = plt.subplots()
ax.margins(0.05)

#Realiza a iteração sobre cada um dos grupos construidos (normal e sobrepeso) x e y, conjunto de dados supervisionado 2d
for name, group in groups:
  ax.plot(group['Comp. Sepalas (cm)'], group['Larg. Sepalas (cm)'], marker='o', linestyle='', ms=12, label=name)

ax.legend(numpoints=1, loc=2)
plt.xlabel('Comp. Sepalas (cm)')
plt.ylabel('Larg. Sepalas (cm)')
plt.title('Comprimento e Largura das Sepalas')
leg = ax.legend(loc = 'lower right')
plt.show()
```

```python
# não supervisionado

df_n_supervisionado.plot.scatter(x='Comp. Sepalas (cm)', y='Larg. Sepalas (cm)', marker='o')
```
