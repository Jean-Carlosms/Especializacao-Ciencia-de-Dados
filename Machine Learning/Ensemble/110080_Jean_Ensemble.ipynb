# **Importando Bibliotecas**

```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_breast_cancer
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.ensemble import AdaBoostClassifier, BaggingClassifier, VotingClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
from sklearn.metrics import classification_report
```

# **Verificando o Dataset e Normalizando os dados**

```python
# Leitura do dataset do data_cancer
data = load_breast_cancer()
X = data['data']
y = data['target']
```

```python
# Normalização dos dados
sc = StandardScaler()
X_norm = sc.fit_transform(X)
```

```python
# Split do Dataset
X_train, X_test, y_train, y_test = train_test_split(X_norm, y, test_size=0.2, random_state=0)
```

# **Adaboost, Bagging, Voting com 3 modelos com Grid Search; melhores parametros, modelos e score**

```python
# Resultado (classification report) com Adaboost
ada = AdaBoostClassifier()
ada.fit(X_train, y_train)
```

```python
y_pred_ada = ada.predict(X_test)
print('Adaboost')
print(classification_report(y_test, y_pred_ada))
```

```python
# Resultado (classification report) com Bagging
bagging = BaggingClassifier()
bagging.fit(X_train, y_train)
```

```python
y_pred_bag = bagging.predict(X_test)
print('Bagging')
print(classification_report(y_test, y_pred_bag))
```

```python
# Definir os classificadores
knn = KNeighborsClassifier()
decision_tree = DecisionTreeClassifier()
svm = SVC()
```

```python
# VotingClassifier com três modelos
voting = VotingClassifier(estimators=[('KNN', KNeighborsClassifier()),
 ('tree', DecisionTreeClassifier()),('SVM', SVC()
  )], voting='hard')
voting.fit(X_train, y_train)
```

```python
y_pred_voting = voting.predict(X_test)
print('Voting')
print(classification_report(y_test, y_pred_voting))
```

```python
# Parametros do Grid Search
param_grid = {
    'KNN__n_neighbors': [3, 5, 7],
    'tree__max_depth': [3, 5, 7, 9],
    'SVM__C': [1, 10, 100],
    'SVM__gamma': [0.1, 1, 10],
    'SVM__kernel': ['linear', 'rbf']
}
```

```python
# Resultado (classification report) do GridSearch do Voting com os modelos
# KNN
# DecisionTrees
# SVM
grid_search = GridSearchCV(voting, param_grid, cv=10, verbose=3)
grid_search.fit(X_train, y_train)
```

```python
y_pred_voting2 = grid_search.predict(X_test)
print('Voting com GridSearch')
print(classification_report(y_test, y_pred_voting))
```

```python
best_params = grid_search.best_params_
print(best_params)
```

```python
best_model = grid_search.best_estimator_
print(best_model)
```

```python
best_score = grid_search.best_score_
print(best_score)
```
