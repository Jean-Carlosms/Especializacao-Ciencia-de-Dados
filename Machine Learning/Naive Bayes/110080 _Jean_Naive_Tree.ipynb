```python
#Importando Bibliotecas

import pandas as pd
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.naive_bayes import GaussianNB
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

# **`Analise Exploratória`**

```python
# Carregando o dataset = breast_cancer
dados = load_breast_cancer()
```

```python
dados
```

```python
# Separando o alvo do dataset
df_dados = pd.DataFrame(data=dados.data, columns=dados.feature_names)
df_dados['target'] = dados.target
```

```python
# Conferindo as informações dos dados
print(df_dados.info())
```

```python
# Verificando se possui algum valor ausente
print("Verificando valores ausentes no dataset:")
print(df_dados.isna().sum())
```

# **Classificando dados**

```python
# Separando as características X e Alvo Y
X = df_dados.drop('target', axis=1)
y = df_dados['target']
```

```python
#Dividindo o dataset de treino e validação
X_treino, X_teste, y_treino, y_teste = train_test_split(X, y, test_size=0.2, random_state=100)
```

```python
# Normalizando os dados
scaler = StandardScaler()
X_treino = scaler.fit_transform(X_treino)
X_teste = scaler.transform(X_teste)
```

```python
# Classification report de Gaussian Naive Bayes
gau = GaussianNB()
gau.fit(X_treino, y_treino)
y_pred = gau.predict(X_teste)
print(classification_report(y_teste, y_pred))
```

```python
#Classification report Decision Tree
tree = DecisionTreeClassifier()
tree.fit(X_treino, y_treino)
y_pred = tree.predict(X_teste)
print(classification_report(y_teste, y_pred))
```
