### 4 - Utilizando funções agregadas:

```python
# Obtendo a média de valores de uma coluna:
df['Unit price'].mean()
```

```python
# Obtendo o valor mínimo de uma coluna:
df['Total'].min()
```

```python
# Obtendo o valor máximo de uma coluna:
df['Total'].max()
```

```python
# Retornando valores no quantil* fornecido sobre o eixo solicitado:
df.quantile(0.25)
```

```python
# Obtendo a moda de alguma coluna (valor que mais aparece):
df['Quantity'].mode()
```

```python
# Agregando mais de uma operação:
df['Rating'].agg([max, min, np.mean])
```

### 5 - Conhecendo outras operações, agrupamentos e slicing:

```python
# Deletando dados duplicados:
df_payments = df.drop_duplicates(subset='Payment')
df_payments
```

```python
# Recuperando a contagem de dados de acordo com as categorias:
df['Product line'].value_counts()
```

```python
# Utilizando o 'normalize' para retornar a frequência de valores únicos:
df[['Gender']].value_counts(normalize = True)
```

```python
# Realizando o agrupamento de 'Unit price' por 'Product line':
df.groupby('Product line')['Unit price'].mean().sort_values(ascending = False)
```
