### 1 - Explorando o dataset:

```python
# Importando o Numpy e Pandas:
import numpy as np
import pandas as pd
```

```python
# Carregando arquivo .csv:
df = pd.read_csv('./supermarket.csv')
```

```python
# Mostrado os 05 primeiros registros do Data Frame:
df.head(5)
```

```python
# Mostrando os 05 últimos registros do Data Frame:
df.tail(5)
```

```python
# Visualizando os detalhes e dados estatísticos do Data Frame:
df.describe()
```

### 2 - Exportando para outros formatos:

```python
# Exportando o Data Frame para outros formatos:
df.to_excel('excel_supermarket.xlsx', sheet_name='Sheet1')
df.to_json('supermarket.json')
```

### 3 - Selecionando os dados e realizando operações:

```python
# Recuperando os dados de uma coluna apenas:
df['City']
```

```python
# Recuperando dados de um grupo de colunas:
df[['City', 'Total']]
```

```python
# Aplicando o describe() apenas para uma coluna:
df['City'].describe()
```

```python
# Visualizado um resumo geral do Data Frame:
df.info()
```

```python
# Visualizando a quantidade de linhas x colunas do Data Frame:
df.shape
```

```python
# Ordenando os dados do Data Frame de acordo com uma das colunas:
df.sort_values('Total')
```

```python
# Visualizado apenas uma das colunas com os valores ordenados (ordem decrescente):
df['City'].sort_values(ascending = False)
```

```python
# Aplicando o filtro por determinado tipo de dado existente em determinada coluna:
df[df['Product line'] == 'Electronic accessories']
```

```python
# Aplicando o filtro com mais de uma condição:
df_yangon_electronics = df[
  (df['City'] == 'Yangon') & (df['Product line'] == 'Electronic accessories')
]

df_yangon_electronics.head()
```

```python
# Criando uma nova coluna e aplicando uma operação matemática entre as variáveis:
df['Ticket'] = df['Total'] / df['Quantity']
df.head()
```

### 4 - Utilizando funções agregadas:

```python
# Obtendo a média de valores de uma coluna:
df["Total"].mean()
```

```python
# Obtendo o valor mínimo de uma coluna:
df["Total"].min()
```

```python
# Obtendo o valor máximo de uma coluna:
df["Total"].max()
```

```python
# Retornando valores no quantil* fornecido sobre o eixo solicitado:
df.quantile(0.25)
```

```python
# Obtendo a moda de alguma coluna (valor que mais aparece):
df["Quantity"].mode()
```

```python
# Agregando mais de uma operação:
df['Rating'].agg([np.mean, max, min])
```

### 5 - Conhecendo outras operações, agrupamentos e slicing:

```python
# Deletando dados duplicados:
df_payments = df.drop_duplicates(subset = 'Payment')
df_payments
```

```python
# Recuperando a contagem de dados de acordo com as categorias:
df['Product line'].value_counts()
```

```python
# Utilizando o 'normalize' para retornar a frequência de valores únicos:
df[['Gender']].value_counts(normalize = True)
```

```python
# Realizando o agrupamento de 'Unit price' por 'Product line':
df.groupby("Product line")["Unit price"].mean().sort_values(ascending = False)
```

```python
# Agrupando com mais de uma variável:
df.groupby("Product line")[["Unit price", "Total"]].mean().sort_values("Total", ascending = False)
```

```python
# Agrupando com múltiplas estatísticas:
df.groupby("Product line")[["Quantity", "Total"]].agg([min, max,sum])
```

```python
# Mostrar as colunas existentes no Data Frame:
df.columns
```

```python
# Realizando o slicing com colunas:
df.loc[:, "City":"Product line"]
```

```python
# Realizando o slicing com posições:
df.iloc[2:10, 1:5]
```

```python
# Validando se existem valores ausentes (NaN = Not a Number):
df.isna().sum()
```

### 6 - Lidando com dados ausentes:

```python
# Criando uma lista por coluna
lista = {
  "nome": ["João", "Maria", "Caio", "Gabriela"],
  "sexo": ["H", "F", "M", "F"],
  "produto": ["abacate", "mamão", "laranja", "laranja"],
  "peso_kg": ["1.2", "2,5", "3.5", np.nan],
  "valor_total": [10.48, 22.5, 14.4, np.nan]
}

# Criando um novo dataframe
df2 = pd.DataFrame(lista)
df2
```

```python
# Total de valores ausentes por coluna
df2.isna().sum()
```

```python
# Plotando valores ausentes
df2.isna().sum().plot(kind="bar")
```

```python
# Removendo valores missing
df2.dropna()
```

```python
# Preenchendo valores ausentes, no caso abaixo com 0:
df2.fillna(0)
```
