Crie uma calculadora utilizando a Linguagem Python, apresentando o seguinte menu interativo para o usuário:

* A - Somar
* B - Subtrair
* C - Multiplicar
* D - Dividir
* E - Potência
* F - Raiz Quadrada
* G - Média
* H - Máximo
* I - Mínimo
* J - Área
* K - Perímetro
* S - Sair


**Considerações importantes sobre a interação com o usuário:**

1 - O menu deverá ser apresentado enquanto o usuário não informar o comando de saída (no caso, representado pela letra 'S');

2 -  Ao término de cada operação escolhida, a tela deverá ser limpa e apresentado novamente o menu principal;

3 -  Caso o usuário informe alguma opção inválida, deverá possibilitar a inserção novamente (até acertar);

4 -  Para as opções 'A', 'B', 'C', 'D', 'G' e 'I' o usuário poderá passar quantos números desejar (caso informe na sequência dos números algum caracter, solicitar a correção ou voltar ao menu principal - disponibilizar essas duas opções, não esquecer de limpar a tela);

5 -  Para a opção 'E', solicitar a base e o expoente. Caso seja informado algum valor indevido, segue mesmas validações do item 4;

6 - Para a opção 'F', idem correções do item 4 (caso seja informado um valor indevido);

7 - Pars os itens 'J' e 'K', solicitar os dados necessários para os cálculos (escolher uma figura geométrica para cada);


**Considerações importantes sobre o desenvolvimento do código:**

- Procure utilizar com propriedade os conceitos aprendidos em sala de aula;
- Reutilizem o código com o uso de funções;



```python
# Código para importar biblioteca para limpar a tela do colab

from IPython.display import clear_output

# Código para importar bibliotéca de operações matemáticas

import math

# Dicionário de operações com funções lambda
operacoes = {
    "A": lambda lista: sum(lista), #Somar
    "B": lambda lista: lista[0] - sum(lista[1:]), # pega o primeiro valor da lista para subtrair com a soma do restante
    "C": lambda lista: math.prod(lista), #Multiplicar
    "D": lambda lista: lista[0] / math.prod(lista[1:]) if 0 not in lista[1:] else "Divisão por zero não é permitida", #Dividir
    "E": lambda lista: lista[0] ** lista[1] if len(lista) == 2 else "A operação de potência requer dois números", #Potência
    "F": lambda lista: [math.sqrt(num) for num in lista], # Raiz Quadrada
    "G": lambda lista: sum(lista) / len(lista), #Média
    "H": lambda lista: max(lista), #Máximo
    "I": lambda lista: min(lista), #Mínimo
    "J": lambda lista: lista[0] ** 2 if len(lista) == 1 else "Área do quadrado requer apenas um número", #Área "QUADRADO"
    "K": lambda lista: 4 * lista[0] if len(lista) == 1 else "Perímetro do quadrado requer apenas um número", #Perímetro "QUADRADO"

}

# Função para visualizar o menu da calculadora e requisitar a letra
def menu_calculadora():
    while True:
        clear_output(wait=True) # limpar após a operação no COLAB
        print("Selecione a operação:")
        print("A- Somar")
        print("B- Subtrair")
        print("C- Multiplicar")
        print("D- Dividir")
        print("E- Potência")
        print("F- Raiz Quadrada")
        print("G- Média")
        print("H- Máximo")
        print("I- Mínimo")
        print("J- Área (Quadrado)")
        print("K- Perímetro (Quadrado)")
        print("S- Sair")

        letra = (input("Digite uma letra do menu: ")).upper() # salva a letra na variável e a transforma em letra maíscula

        if (letra == "A" or letra == "B" or letra == "C" or letra == "D" or letra == "E" or letra == "F" or
        letra == "G" or letra == "H" or letra == "I"):
            entrada = input("Digite os números para realizar o calculo desejado: ")
            numeros = list(map(float, entrada.split())) #divide a entrada em várias minis entradas e depois coverter todas para float
            resultado = operacoes[letra](numeros) # função condicional para calcular as lambdas que estão no dicionário com os números digitados para calculo
            print(f"Resultado: {resultado}")
            input("Digite uma tecla, para continuar com o programa... ")  # Pausa até digitar uma tecla

        elif (letra == "J"):
            entrada = input("Digite um número para calcular a área do quadrado: ")
            numeros = list(map(float, entrada.split())) #divide a entrada em várias minis entradas e depois coverter todas para float
            resultado = operacoes[letra](numeros) # função condicional para calcular as lambdas que estão no dicionário com os números digitados para calculo
            print(f"Resultado: {resultado}")
            input("Digite uma tecla, para continuar com o programa... ")  # Pausa até digitar uma tecla

        elif (letra == "K"):
            entrada = input("Digite um número para calcular o perímetro do quadrado: ")
            numeros = list(map(float, entrada.split())) #divide a entrada em várias minis entradas e depois coverter todas para float
            resultado = operacoes[letra](numeros) # função condicional para calcular as lambdas que estão no dicionário com os números digitados para calculo
            print(f"Resultado: {resultado}")
            input("Digite uma tecla, para continuar com o programa ... ")  # Pausa até digitar uma tecla
        elif letra == "S" :
            print("Saindo da calculadora")
            break # sai do programa
        else:
            print("Letra fora do menu ")

# Executar a calculadora
menu_calculadora()
```
