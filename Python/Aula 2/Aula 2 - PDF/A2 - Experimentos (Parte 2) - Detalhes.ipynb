# Estruturas de Fluxo (Repetição com For)

Iteração em qualquer tipo de sequência. Abaixo, um exemplo de repetição com base numa string:

```python
for character in 'hello':
    print(character)
```

Iteração em um intervalo definido (**min_value**, **max_value**):



```python
for i in range(5, 8):
    print(i, i ** 2)
```

Quando é utilizado apenas um parâmetro, considera-se que o **min_value** é 0 (zero). Lembrando que sempre o max_value será -1:

```python
for i in range(3):
    print(i)
```

Possibilidade de criar a regra de repetição a partir de algum cálculo:




```python
for i in range(2 ** 2):
    print('Hello, world!')
```

Exemplo de iteração não realizada:

```python
for i in range(-5):
    print('Hello, world!')
```

No exemplo abaixo, o intervalo será a partir de 1 até n + 1 (ou seja, o max_value inclusive):




```python
result = 0
n = 5
for i in range(1, n + 1):
    result += i
    # este ^^ é a abreviação de
    # resultado = resultado + i
print(result)
```

É possível também informar três parâmetros (**start_value**, **end_value**, **step**). O exemplo abaixo mostra uma iteração na ordem decrescente, de 2 em 2:

```python
for i in range(10, 0, -2):
    print(i)
```

Para realizar a iteração seguindo o príncipio de percorrer cada item de dado conjunto de dados, podemos utilizar o **in**:

```python
example_list = [1, 2, 3, 4, 5, 6]

for n in example_list:
  print(n)
```

Execução encadeada + instruções de controle (*break*, *pass*, continue):

```python
for x in range(3):
  for y in range(3):
    if(y == 1):
      print('Pulou o 1!')
      continue

    print(x, ' e ', y)
```

```python
for x in range(5):
  if x == 2:
    pass
  else:
    print(x)

  print('Continuando o código na mesma iteração')
```

```python
# Outros exemplos:

# BREAK (encerra o laço de repetição corrente):
print('\nBREAK:')
for x in range(5):
    if x == 3:
        break

    print(x)

# CONTINUE (encerra o laço de repetição corrente e inicializa o próximo, se houver):
print('\nCONTINUE:')
for x in range(5):
    if x == 3:
        continue

    print(x)


# PASS (simplesmente pula determinada condição, mas continua na mesma iteração):
print('\nCONTINUE:')
for x in range(5):
    if x == 3:
        pass
    else:
        print(x)

    print('Quando pular o número 3, continuarei sendo executado na mesma iteração!')
```

# Estruturas de Fluxo (Repetição com While)

```python
count = 5

while count >= 0:
  print(count)

  count -= 1
```
