```python
# Definindo uma função sem retorno:
def mostrar_frase():
    print('Olá, seja bem-vindo!')

mostrar_frase()
```

```python
# Definindo uma função com retorno:
def calcular():
    return 4 ** 4

calcular()
```

```python
# Definindo uma função com parâmetros e retorno:
def somar(n1, n2):
    return n1 + n2

somar(2, 2)
```

```python
# Definindo uma função com dois retornos:
def recuperar_nome_idade():
    return "Adriano", 31

nome, idade = recuperar_nome_idade()
print(f'Nome: {nome} - Idade: {idade}')
```

```python
# Definindo uma função com parâmetros indefinidos:
def somar(*params):
  print(params[0])
  print(params[1])

somar(1, 2)

# Os parâmetros indefinidos sempre devem ser transmitidos no final de todos os demais (se existirem)!
def apresentar(nome, idade, *desejos):
  print(f'Meu nome é {nome} e tenho {idade} anos!')
  print(f'Meus desejos são: {desejos}')

apresentar("Adriano", 31, "Informática", "Carros", "Mulheres")
```
