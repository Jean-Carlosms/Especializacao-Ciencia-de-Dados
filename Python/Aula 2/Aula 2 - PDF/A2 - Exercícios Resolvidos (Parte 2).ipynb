```python
# Exercício 03:
number = -17,3

# number = input('Input int number: ')
# number = int(number) if number.isnumeric() else  ""

if(type(number) == int and number > 0):
    maximum_value = 5.5 * (10**7)

    if(number <= maximum_value):
        n = 1
        prime = 0

        while(n <= number):
            if(number % n == 0):
                prime += 1

            n += 1

        if(prime == 2):
            print('É primo!')
        else:
            print('Não é primo!')

    else:
        print('Número fora do intervalo permitido!')

else:
    print('Número Inválido!')
```
