```python
# Exercício 01:
import pprint

numbers = [6, 17, 9, 8, 1, 12]

evens = []
odds = []

evens.append(numbers[0]) if numbers[0] % 2 == 0 else odds.append(numbers[0])
evens.append(numbers[1]) if numbers[1] % 2 == 0 else odds.append(numbers[1])
evens.append(numbers[2]) if numbers[2] % 2 == 0 else odds.append(numbers[2])
evens.append(numbers[3]) if numbers[3] % 2 == 0 else odds.append(numbers[3])
evens.append(numbers[4]) if numbers[4] % 2 == 0 else odds.append(numbers[4])
evens.append(numbers[5]) if numbers[5] % 2 == 0 else odds.append(numbers[5])

t = (evens, odds)

pprint.pprint(t)
```

```python
# Exercício 02:
import pprint

name = input('Informe o nome: ')

age = int(input('Informe a idade: '))

first_interest = input('Informe o primeiro interesse: ')
second_interest = input('Informe o segundo interesse: ')

interests = []
interests.append(first_interest)
interests.append(second_interest)

dictionary = {
    'nome': name,
    'idade': age, 
    'interesses': interests, 
    'informacoes_adicionais': {
        'quantidade_letras_do_nome': len(name), 
        'idade_ao_quadrado': pow(age, 2), # age ** 2
        'lista_dos_numeros_da_idade': list(str(age)) # [str(age)[0], str(age)[1]]
    }
}

print(dictionary)
pprint.pprint(dictionary)
```
