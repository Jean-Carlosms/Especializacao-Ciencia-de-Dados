```python
# Exercício 06: 
def growth_rate(population_a, population_b):
    years = 0
    response = ''
    wrong = 0
    
    if(type(population_a) != int or type(population_b) != int):
        wrong += 1
    
    elif(population_a < 0 or population_b < 0):
        wrong += 1
    
    if(wrong != 0):
        response = 'invalid argument'
    
    else:
        while(population_a < population_b):
            population_a += (population_a * 3) / 100
            population_b += (population_b * 1.5) / 100
        
            years += 1
        
        response = years
    
    return response
```

```python
# Exercício 07: 
def sum_of_products(list_one, list_two):
    wrong_number = 0
    response = ''
    sum_lists = 0
    max_value = 0
    
    for number in list_one:
        if(type(number) != int and type(number) != float):
            wrong_number += 1
    
    for number in list_two:
        if(type(number) != int and type(number) != float):
            wrong_number += 1
      
    if(wrong_number != 0):
        response = 'wrong number'
    
    if(wrong_number == 0): 
        if(len(list_one) == 0 and len(list_two) == 0):
            response = -1
            
        elif(len(list_one) == len(list_two)):
            n = 0

            while(n < len(list_one)):
                sum_lists += list_one[n] * list_two[n]
                n += 1

            response = sum_lists     

        else:
            if(len(list_one) > len(list_two)): 
                max_value = len(list_one) 

            elif(len(list_one) < len(list_two)):
                max_value = len(list_two)

            if(max_value > 0):
                n = 0

                while(n < max_value):
                    if(len(list_one) == 0 or len(list_one) <= n):
                        sum_lists += list_two[n] * 1

                    elif(len(list_two) == 0 or len(list_two) <= n):
                        sum_lists += list_one[n] * 1

                    elif(len(list_one) >= n and len(list_two) >= n):
                        sum_lists += list_one[n] * list_two[n]

                    n += 1

                response = sum_lists
    
    return response
```

```python
list_1 = [1, 2, 1, 1]
list_2 = [1, 2, 3, 4]

sum_of_products(list_1, list_2)
```

```python
# Exercício 08: 
def countlist(numbers):  
    max_value = 0
    sum_values = 0
    occurs_value = 0
    mean_values = 0
    near_mean_value = 0
    minus_sum_values = 0
    
    count_values = 0
    dictionary = {}
    
    for number in numbers:
        count_values += 1
        
        #sum
        sum_values += number
        
        #max
        if(count_values == 1):
            max_value = number
            
        elif(number > max_value):
            max_value = number
        
        #occurs
        if(number == numbers[0]):
            occurs_value += 1
        
        #mean
        mean_values = sum_values / count_values
        
        #near-mean
        less_difference = 0
        
        
        for n in numbers:
            if(n == mean_values):
                near_mean_value = n
                break
                
            difference = mean_values - n

            if(difference < 0):
                difference = difference * -1

            if(less_difference == 0 or less_difference > difference):
                near_mean_value = n
                less_difference = difference
        
        
        #minus
        if(number < 0):
            minus_sum_values += number
    
    
    dictionary = {    
        'max' : max_value, 
        'sum' : sum_values,
        'occurs' : occurs_value,
        'mean' : mean_values,
        'near-mean' : near_mean_value,
        'minus' : minus_sum_values}
        
    return dictionary.items()
```

```python
numbers = [1, 2, 3, 4, 5, -7, -8, 9]
countlist(numbers)
```
