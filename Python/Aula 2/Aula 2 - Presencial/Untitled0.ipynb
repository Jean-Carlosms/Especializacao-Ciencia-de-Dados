```python

def growth_rate(population_a, population_b):
    if type(population_a) != int or type(population_b) != int:
      print("Invalid argument")
    elif population_a < 0 or population_b < 0:
      print("Invalid argument")

    else:
      years = 0
      while population_a < population_b:
        population_a += int(population_a * 0.03)
        population_b += int(population_b * 0.015)
        years += 1

      return years

population_a = 90000
population_b = 200000
result = growth_rate(population_a, population_b)

if result is not None:
  print(f"em {result} na população A passará a População B")
```
