```python
# Função com Parâmetros e com retorno
def somar(n1, n2):
  return n1 + n2

r = somar(2, 3)
print(r)
```

```python
 # Keys Params Python pesquisar* def somar_numeros(**params):
def somar_numeros(*params):
  for p in params:
    print(p)
  print(params)
somar_numeros(2, 3, 4, 5, 6, 7)
```

```python
lista = [1, 2, 3, 4, 5]
def compare(n):
  if n % 2 == 0:
    return n
r = list(map(compare, lista))
print(r)
```

```python
lista = [1, 2, 3, 4, 5]
def compare(n):
  if n % 2 == 0:
    return n

r = list(map(compare, lista))
print(r)
```

```python
lista = [1, 2, 3, 4, 5]
def compare(n):
  if n % 2 == 0:
    return n
r = list(filter(compare, lista))
print(r)
```
