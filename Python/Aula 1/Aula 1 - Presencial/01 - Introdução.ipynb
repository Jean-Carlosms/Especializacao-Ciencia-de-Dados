```python
#Modificando o tipo da variável  "STR, INT, FLOAT"
nome = input("Entre com o seu nome:")
idade = input ("Informe a sua idade:")

type(nome)
nome = 2
print(type(nome))
```

```python
# Mostra o Tipo e a Posição da memória
# nome, tipo, valor e posição de memória
nome = "Jean"
print(type(nome))
print(id(nome))

nome = 3
print(type(nome))
print(id(nome))
```

```python
# Casting converte as variáveis"força a cocatenação de STR com INT"
# Interpolação acrescentar variáveis dentro do texto
print("Olá, meu nome é " + str(nome))

print(f"Olá, meu nome é {nome}")
print("Olá, meu nome é {0}".format(nome))
```
