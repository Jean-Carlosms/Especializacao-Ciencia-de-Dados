```python
# Exercício 7
a = input("Informe o Variável A ")
b = input("Informe o Variável B ")
c = input("Informe o Variável C ")
delta =  pow(float(b),2) - 4*float(a)*float(c)
raiz_delta = pow(float(delta),0.5)

if delta >= 0 :

  print(f"X1 = {((-(float(b)) + raiz_delta) / (2 * float(a)))}")
  print(f"X2 = {((-(float(b)) - raiz_delta) / (2 * float(a)))}")

else:
  print("O delta é negativo. Equação não possui raízes reais.")
```
