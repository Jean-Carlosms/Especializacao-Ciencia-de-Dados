```python
# exercício 2
nbase= input("Digite o número base:")
nexpo= input("Digite o número expoente:")
resultado = int(nbase)**int(nexpo)
print(f"O número {nbase} elevado a {nexpo} resulta em {resultado}")
```

```python
# Exercício 03:
print('Informe uma frase: ')
frase = input()

print('Informe uma palavra existente na frase: ')
palavra_existente = input()

print('Informe a palavra que deseja substituir: ')
palavra_substituir = input()

nova_frase = frase.replace(
    palavra_existente,
    palavra_substituir.upper())

print(nova_frase)
```
