```python
# Exercício 01:
print('Informe o nome: ')
nome = input()

print('Informe a idade: ')
idade = input()

print(f'Bem-vindo {nome}. Você nasceu em {2023 - int(idade)}')
```

```python
# Exercício 02:
print('Informe a base: ')
base = input()

print('Informe o expoente: ')
expoente = input()

resultado = pow(int(base), int(expoente))

print(f'O número {base} elevado a {expoente} resulta em {resultado}')
```

```python
# Exercício 03:
print('Informe uma frase: ')
frase = input()

print('Informe uma palavra existente na frase: ')
palavra_existente = input()

print('Informe a palavra que deseja substituir: ')
palavra_substituir = input()

nova_frase = frase.replace(
    palavra_existente,
    palavra_substituir.upper())

print(nova_frase)
```
