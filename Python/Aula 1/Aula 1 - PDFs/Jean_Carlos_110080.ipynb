1 - Escreva um algoritmo que leia um número inteiro qualquer do usuário, e informe se o mesmo é par ou ímpar. Caso o usuário informe um número real, informar a mensagem "O número não é inteiro.". Caso seja informado um número negativo, informar a mensagem "O número deve ser positivo.".

```python
número= float(input("Informe um número "))

if(número == número // 1):

  if(número >= 0):

    if(número % 2 == 0 ):
      print("Número é par")
    else:
      print("Número é ímpar")

  else:
    print("O número deve ser positivo")

else:
  print("O número não é inteiro")
```

2 - Escreva um algoritmo que leia três valores inteiros e diferentes, mostrando-os no final na ordem decrescente. Caso existam valores iguais, informar a mensagem "Foram informados números repetidos.".

```python
valor1= float(input("Informe o primeiro valor inteiro "))
valor2= float(input("Informe o segundo valor inteiro "))
valor3= float(input("Informe o terceiro valor inteiro "))

if((valor1 == valor1 // 1) and (valor2 == valor2 // 1) and (valor3 == valor3 // 1)):

  if(valor1 != valor2 and valor1 != valor3 and valor2 != valor3):

    if(valor1 > valor2 and valor1 > valor3 and valor2 > valor3):
      print(f"{valor1}, {valor2}, {valor3}")
    elif(valor2 > valor1 and valor2 > valor3 and valor1 > valor3):
      print(f"{valor2}, {valor1}, {valor3}")
    elif(valor3 > valor1 and valor3 > valor2 and valor1 > valor2):
      print(f"{valor3}, {valor1}, {valor2}")
    elif(valor1 > valor2 and valor1 > valor3 and valor3 > valor2):
      print(f"{valor1}, {valor3}, {valor2}")
    elif(valor2 > valor1 and valor2 > valor3 and valor3 > valor1):
      print(f"{valor2}, {valor3}, {valor1}")
    else:
      print(f"{valor3}, {valor2}, {valor1}")

  else:
    print("Foram informados números repetidos")

else:
  print("Informar apenas números inteiros")
```

3 - Escreva um algoritmo que calcule o valor a ser pago por um determinado produto, considerando as seguintes condições:

* A: à vista com 10% de desconto;
* B: em duas vezes, sem juros;
* C: em três vezes, com 5% de juros (acima do valor original).


Para isso, devem ser consideradas duas entradas do usuário: valor do produto e condição de pagamento (A, B ou C). Como saída, deverá ser informado o valor a ser pago com os reajustes (no caso de parcelas, informar o valor de cada parcela).



```python
valor_produto= float(input("Informe o valor do produto "))
tipo_pagamento= str(input("Informe o método de pagamento "))

if(tipo_pagamento == "A" or tipo_pagamento == "a" ):
  print(f"A vista será com 10% de desconto e o valor total será de R$ {valor_produto*0.90}")

elif(tipo_pagamento == "B" or tipo_pagamento == "b"):
  print(f"O valor do produto é R$ {valor_produto} e será dívido em 2 x R$ {valor_produto/2} sem juros")

elif(tipo_pagamento == "C" or tipo_pagamento == "c"):
  print(f"O valor total do produto com 5% de juros será de R$ {valor_produto*1.05} e será dívido em 3 x R$ {valor_produto/3*1.05} ")

else:
  print("Condição de pagamento informado está errado")
```
