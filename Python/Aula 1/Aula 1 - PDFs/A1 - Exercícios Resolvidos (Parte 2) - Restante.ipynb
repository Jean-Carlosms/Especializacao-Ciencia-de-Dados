```python
# Exercício 07:
from math import *

print('Informe o valor de A: ')
a = input()

print('Informe o valor de B:')
b = input()

print('Informe o valor de C:')
c = input()

delta = pow(float(b), 2) - (4 * float(a) * float(c))

if delta <= 0:
  print("Impossível realizar a operação")
else:
  x_i = (- float(b) + sqrt(delta)) / (2 * float(a))
  x_ii = (- float(b) - sqrt(delta)) / (2 * float(a))

  print(f'XI: {round(x_i, 2)}')
  print(f'XII: {round(x_ii, 2)}')
```

```python
# Exercício 08:

print('Informe o primeiro lado: ')
lado_um = float(input())

print('Informe o segundo lado: ')
lado_dois = float(input())

print('Informe o terceiro lado: ')
lado_tres = float(input())

if(lado_um == lado_dois == lado_tres):
  print('O triânculo é equilátero!')
elif((lado_um == lado_dois) or (lado_um == lado_tres) or (lado_dois == lado_tres)):
  print('O triângulo é isósceles!')
else:
  print('O triângulo é escaleno!')
```
