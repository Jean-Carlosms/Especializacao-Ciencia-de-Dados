```python
# Exercício 04:

print('Informe o nome: ')
nome = input()

print('Informe a nota: ')
nota = input()

situacao = ''

if(float(nota) >= 0 and float(nota) < 4):
  situacao = 'REPROVADO'
elif(float(nota) >= 4 and float(nota) < 7 ):
  situacao = 'EM RECUPERAÇÃO'
elif(float(nota) >= 7 and float(nota) <= 10 ):
  situacao = 'APROVADO'
else: 
  situacao = 'INVÁLIDO'

if(situacao == 'INVÁLIDO'):
  print('Nota informada inválida!')
else:
  print(f'Olá {nome}, você está {situacao}')
```

```python
# Exercício 05:

print('Informe o salário bruto: ')
salario_bruto = input()

print('Informe o tempo de empresa: ')
tempo_empresa = input()

print('Utiliza VT?')
vale_transporte = input()

salario_liquido = 0.00
inss = 0.00
desconto_vale_transporte = 0.00
adicional_tempo_empresa = 0.00

if(float(salario_bruto) <= 1200.00):
  inss = float(salario_bruto) * 0.07
elif(float(salario_bruto) >= 1200.01 and float(salario_bruto) <= 5000.00):
  inss = float(salario_bruto) * 0.13
else:
  inss = float(salario_bruto) * 0.16


desconto_vale_transporte = float(salario_bruto) * 0.06 if vale_transporte.upper() == 'S' else 0.00

if(int(tempo_empresa) == 1):
  adicional_tempo_empresa = float(salario_bruto) * 0.02
elif(int(tempo_empresa) >= 2 and int(tempo_empresa) <= 4):
  adicional_tempo_empresa = float(salario_bruto) * 0.05
elif(int(tempo_empresa) >= 5 and int(tempo_empresa) <= 10):
  adicional_tempo_empresa = float(salario_bruto) * 0.10
elif(int(tempo_empresa) >= 10):
  adicional_tempo_empresa = float(salario_bruto) * 0.20

salario_liquido = float(salario_bruto) - inss - desconto_vale_transporte + adicional_tempo_empresa

print(f'O seu salário líquido calculado foi de R$ {round(salario_liquido, 2)}')
```
