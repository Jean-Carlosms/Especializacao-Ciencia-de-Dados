* operation -> aeronave_operador_categoria

* manufacturer -> aeronave_fabricante
* engine_type aeronave_motor_tipo
* engines -> aeronave_motor_quantidade
* year_manufacturing -> aeronave_ano_fabricacao
* seating -> aeronave_assentos
* fatalities -> total_fatalidades

Considerando apenas acidentes cujo nível de dano da aeronave tenha sido LEVE ou NENHUM (coluna aeronave_nivel_dano) E que o número de fatalidades (total_fatalidades) tenha sido superior à 0 (zero).

````python
'''
  [x] Criar uma função que faça a leitura e processamento dos arquivos
  [x] Ler o arquivo
  [x] Realizar a o agrupamento por fase de operação
  [x] Calcular a quantidade de ocorrências que cada fase de operação terá no conjunto de dados
  [x] Calcular o percentual que essa quantidade representa ao conjunto total de registros
  [x] Salvar em um arquivo .json com nome statistics.json
  [x] Salvar em um arquivo .CSV com nome levels.csv contendo as informações requisitadas
  [x] Considerar apenas os acidentes que tenham sido leve ou nenhum
  [x] E que o número de fatalidades tenha sido superior a 0
  [x] Efetuar apenas uma leitura do arquivo de entrada

'''
````

````python
import csv, json



def salvar_json(dados):
  with open('/content/statistics.json', 'w', encoding= 'UTF-8') as arquivo:
    json.dump(dados, arquivo, ensure_ascii=False, indent=4)

def calcular_percentual(dados, quantidade_linhas):
  for dado in dados:
    dado['percentual'] = str(round(float(dado['ocorrencias'] * 100 / quantidade_linhas), 2)).replace('.',',') + '%'

  return dados

def processar_arquivo():
  conjunto_dados = []
  quantidade_linhas = 0

  with open('/content/anv.csv', 'r', encoding='UTF-8') as arquivo:
    dados = csv.reader(arquivo, delimiter = '~')
    next(dados)

    for linha in dados:
      quantidade_linhas += 1
      encontrado = False

      for dado in conjunto_dados:
        if dado["fase_operacao"] == linha [19]:
          dado['ocorrencias'] += 1

          encontrado = True
          break

      if encontrado == False:
        conjunto_dados.append({
            "fase_operacao": linha[19],
            "ocorrencias": 1,
            "percentual": 0.00
        })

  conjunto_dados = calcular_percentual(conjunto_dados, quantidade_linhas)
  salvar_json(conjunto_dados)

````

````python
processar_arquivo()
````

````python
'''
[x]Crie um arquivo de saida (formato CSV) com nome levels.csv contendo as seguintes informações:

operation -> aeronave_operador_categoria
type -> aeronave_tipo_veiculo
manufacturer -> aeronave_fabricante
engine_type aeronave_motor_tipo
engines -> aeronave_motor_quantidade
year_manufacturing -> aeronave_ano_fabricacao
seating -> aeronave_assentos
fatalities -> total_fatalidades

[x]Considerando apenas acidentes cujo nível de dano da aeronave tenha sido LEVE ou NENHUM (coluna aeronave_nivel_dano) E que o número de fatalidades (total_fatalidades) tenha sido superior à 0 (zero).

'''


import csv

with open('/content/anv.csv', 'r', newline='', encoding='UTF-8') as arquivo:   # lendo arquivo ans.csv e usando delimitador '~'
  dados_dano = csv.DictReader(arquivo, delimiter = '~')
  next(dados_dano)
  dados = []


  for linha in dados_dano:
    if linha ['aeronave_nivel_dano'] in ['LEVE', 'NENHUM'] and int(linha['total_fatalidades']) > 0 :   # separador o contéudo requisitado ("LEVE, NENHUM") por dicionário
      dicionario = {
                'operation': linha['aeronave_operador_categoria'],
                'type': linha['aeronave_tipo_veiculo'],
                'manufacturer': linha['aeronave_fabricante'],
                'engine_type': linha['aeronave_motor_tipo'],
                'engines': linha['aeronave_motor_quantidade'],
                'year_manufacturing': linha['aeronave_ano_fabricacao'],
                'seating': linha['aeronave_assentos'],
                'fatalities': linha['total_fatalidades']
            }
      dados.append(dicionario)

with open('/content/levels.csv', 'w', newline='', encoding='UTF-8') as arquivo:      # criando arquivo levels.csv  com as colunas específicas
    colunas = ['operation', 'type', 'manufacturer', 'engine_type', 'engines', 'year_manufacturing', 'seating', 'fatalities']

    writer = csv.DictWriter(arquivo, colunas)
    writer.writeheader()
    writer.writerows(dados)


````

## Exercício 2

Crie um programa que efetue a leitura do arquivo produzido no Exercício 1 (levels.csv) e mostre na tela as seguintes informações estatísticas:

* Quantidade total de acidentes.
* Quantidade total de acidentes agrupados por tipo de aeronave (type).
* Ano e fabricante da aeronave mais antiga (year_manufacturing, manufacturer).
* Quantidade de assentos e tipo de motor da aeronave mais nova (seating, engine_type).
* A quantidade de acidentes que ocorreram com aeronaves do tipo (type)HELICÓPTERO, cuja fabricação se deu após o ano 1997.
* A quantidade de acidentes que ocorreram com aeronaves do tipo (type) AVIÃO fabricadas (manufacturer) pela CESSNA AIRCRAFT.

**Observações**:
1. Efetuar todas as leituras e gravações de arquivos na pasta /tmp

````python
'''
[x]Crie um programa que efetue a leitura do arquivo produzido no Exercício 1 (levels.csv)

 mostre na tela as seguintes informações estatísticas:
[x]Quantidade total de acidentes.
[x]Quantidade total de acidentes agrupados por tipo de aeronave (type).
[x]Ano e fabricante da aeronave mais antiga (year_manufacturing, manufacturer).
[x]Quantidade de assentos e tipo de motor da aeronave mais nova (seating, engine_type).
[x]A quantidade de acidentes que ocorreram com aeronaves do tipo (type)HELICÓPTERO, cuja fabricação se deu após o ano 1997.
[x]A quantidade de acidentes que ocorreram com aeronaves do tipo (type) AVIÃO fabricadas (manufacturer) pela CESSNA AIRCRAFT.
'''

import csv


with open('/content/levels.csv', 'r', encoding='UTF-8') as arquivo: # lendo arquivo levels.csv
    dados_acidentes = csv.DictReader(arquivo)

    total_acidentes = 0
    acidentes_por_tipo = {}
    ano_fabricante_antiga = 2023
    assentos_motor_nova = 0
    acidentes_helicoptero_1997 = 0
    acidentes_cessna_aircraft = 0

    for linha in dados_acidentes:
        total_acidentes += 1


        if linha['type'] in acidentes_por_tipo:   # Acidentes por tipo de aeronave
          acidentes_por_tipo[linha['type']] += 1
        else:
          acidentes_por_tipo[linha['type']] = 1


        if int(linha['year_manufacturing']) < ano_fabricante_antiga:     # Ano e o fabricante mais antigo
            ano_fabricante_antiga = int(linha['year_manufacturing'])
            fabricante_antiga = linha['manufacturer']


        if int(linha['year_manufacturing']) > ano_fabricante_antiga:     # Assento e motor da aeronava mais nova
            assentos_motor_nova = (linha['seating'], linha['engine_type'])


        if linha['type'] == 'HELICÓPTERO' and int(linha['year_manufacturing']) > 1997:       # Helicóptero fabricado após 1997
            acidentes_helicoptero_1997 += 1


        if linha['type'] == 'AVIÃO' and linha['manufacturer'] == 'CESSNA AIRCRAFT':  # Acidentes pela empresa CESSNA AIRCRAFT
            acidentes_cessna_aircraft += 1


print(f"Quantidade total de acidentes: {total_acidentes}")    # Impressões
print(f"Quantidade de acidentes agrupados por tipo de aeronave e modelo: {acidentes_por_tipo}")
print(f"Ano e fabricante da aeronave mais antiga: ({ano_fabricante_antiga}, {fabricante_antiga})")
print(f"Quantidade de assentos e tipo de motor da aeronave mais nova: {assentos_motor_nova}")
print(f"Quantidade de acidentes com helicópteros fabricados após 1997: {acidentes_helicoptero_1997}")
print(f"Quantidade de acidentes com aviões fabricados pela CESSNA AIRCRAFT: {acidentes_cessna_aircraft}")

````
