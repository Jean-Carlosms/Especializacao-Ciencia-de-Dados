```python
# Método construtor, atributos inicializados no construtor
#visibilidade
class Veiculo():
  def __init__(self):
    self.marca = '' # __ váriavel apenas visivel dentro da class
    self.modelo = ''

  def mostrar_detalhes(self):
    print(f"Marca: {self.marca} e modelo: {self.modelo}")

v = Veiculo()
v.marca = "Hyundai"
v.modelo = "HB20"
v.mostrar_detalhes()
```

```python
class Residencia():
  def __init__(self, tipo, comodos):
    self.tipo = tipo
    self.comodos = comodos

#defool method str
  def __str__(self):
    return 'Essa classe simboliza uma residência'

  def apresentar(self):
    print(f'Tipo: {self.tipo}, Cômodos: {self.comodos} ')

r = Residencia('Casa',30)

print(r)
```

```python
class Pessoa():
  def __init__(self):
    self.__nome = 'Jean'
    self.__idade = 0

  #pesquisar sobre "decorators ou annotations"

  @property # encapsulamento = proteger uma parte do seu código
  def nome(self):
    return self.__nome

  @nome.setter # seder o atributo privado para ler essa variavel e escrever nessa váriavel
  def nome(self, nome):
    self.__nome = nome

p = Pessoa()
p.nome = 'dadadadasdasda'
```

```python
class PessoaA():
  def apresentar(self):
    print('Apresentando a pessoa A')

class PessoaB(PessoaA):
  def apresentar(self):
    print('Apresentando a pessoa B')

class PessoaC(PessoaB):
  pass

class PessoaX():
  def __str__(self):
    return 'Apresentando a pessoa X'

class PessoaY(PessoaX):
  def __str__(self):
    return super().__str__() + 'ops ... pessoa Y'


pessoa_x = PessoaX()
print(pessoa_x)

pessoa_y = PessoaY()
print(pessoa_y)

#pessoa_b = PessoaB()
#pessoa_b.apresentar()
#pessoa_c = PessoaC()
#pessoa_c.apresentar()
```
