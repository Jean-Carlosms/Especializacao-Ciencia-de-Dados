### 3.1.1 Gráficos de Linhas

#### Os gráficos de linhas são utilizados para exibir tendências ao longo do tempo ou dados contínuos . Eles são uma ferramenta fundamental para visualizar mudanças e relações em dados sequenciais, tornando-os um ponto de partida crucial para o treinamento. A simplicidade com que os gráficos de linhas representam a evolução dos dados os torna ideais para transmitir informações sobre progressão, flutuações e correlações ao longo de um período ou em relação a uma variável contínua.  


````python
import matplotlib.pyplot as plt

# Dados de amostra
x = [1, 2, 3, 4, 5]
y = [2, 4, 1, 5, 3]

# Cria o gráfico de linhas
plt.plot(x, y)

# Exibe o gráfico
plt.show()
````

#### É possível plotar múltiplas linhas no mesmo gráfico chamando a função plt.plot() diversas vezes dentro do mesmo script. Isso permite comparar diferentes conjuntos de dados ou visualizar múltiplas variáveis em relação ao mesmo eixo.

````python
import matplotlib.pyplot as plt

# Dados de amostra para a primeira linha
x1 = [1, 2, 3, 4, 5]
y1 = [2, 4, 1, 5, 3]

# Dados de amostra para a segunda linha
x2 = [1, 2, 3, 4, 5]
y2 = [3, 1, 4, 2, 5]

# Plota a primeira linha
plt.plot(x1, y1, label='Linha 1')

# Plota a segunda linha
plt.plot(x2, y2, label='Linha 2')

# Adiciona uma legenda para identificar as linhas
plt.legend()

# Exibe o gráfico
plt.show()
````

#### A Matplotlib oferece diversas opções para personalizar a aparência dos gráficos de linhas como por exemplo cores, estilos de linha e marcadores

#### **Cores**: A cor das linhas pode ser alterada utilizando o parâmetro color.


```
    plt.plot(x, y, color='red')
    plt.plot(x, y, color='g')
```

  
#### Cores podem ser especificadas por nomes (como 'red', 'green', 'blue'), códigos hexadecimais (como '#00FF00'), tuplas RGB ou RGBA (como (0, 1, 0, 1)) ou intensidades de escala de cinza (como '0.8').

#### Por exemplo, para plotar uma linha azul, você pode usar color='blue' ou simplesmente 'b'.

````python
import matplotlib.pyplot as plt

x = [4, 5, 6, 7, 8]
y1 = [5, 7, 4, 8, 6]
y2 = [6, 4, 7, 5, 8]

plt.plot(x, y1, color='red', label='Linha Vermelha')
plt.plot(x, y2, color='green', label='Linha Verde')
plt.legend()
plt.show()
````

#### **Estilos de Linha**: O parâmetro linestyle permite modificar o estilo da linha, como tracejado ('--') ou pontilhado (':').

#### Outros estilos incluem linha sólida ('-'), traço-ponto ('-.') e nenhum estilo ('' ou ' ').

#### É possível especificar um padrão de traço personalizado usando uma tupla de dois números: o comprimento do traço e o comprimento da lacuna em pontos.

````python
import matplotlib.pyplot as plt

x = [4, 5, 6, 7, 8]
y = [5, 7, 4, 8, 6]

plt.plot(x, y, linestyle='--', label='Tracejado')
plt.plot(x, [y_val + 1 for y_val in y], linestyle=':', label='Pontilhado')
plt.plot(x, [y_val + 2 for y_val in y], linestyle='-.', label='Traço-Ponto')
plt.plot(x, [y_val + 3 for y_val in y], linestyle='-', label='Sólido')
plt.legend()
plt.show()
````

#### **Marcadores**: Para destacar pontos de dados individuais, o parâmetro marker pode ser utilizado com diferentes símbolos, como círculos ('o') ou asteriscos ('*'). 

#### Existem diversos marcadores disponíveis, incluindo pontos ('.'), vírgulas (,), círculos ('o'), triângulos ('v', '^', '<', '>'), quadrados ('s'), pentágonos ('p'), estrelas ('*', 'h', 'H'), cruzes ('+', 'x', 'D', 'd'), linhas verticais ('|') e linhas horizontais ('_'). 

#### Também é possível personalizar a cor da face do marcador (markerfacecolor), a cor da borda (markeredgecolor) e o tamanho (markersize).

````python
import matplotlib.pyplot as plt

x = [4, 5, 6, 7, 8]
y = [5, 7, 4, 8, 6]

plt.plot(x, y, marker='o', label='Círculos')
plt.plot(x, [y_val + 1 for y_val in y], marker='*', label='Asteriscos')
plt.plot(x, [y_val + 2 for y_val in y], marker='s', label='Quadrados')
plt.legend()
plt.show()
````

#### É possível combinar a especificação de cor, estilo de linha e marcador em uma única string de formato. A ordem dos elementos na string de formato é marcador, estilo de linha e cor. Por exemplo, 'ro--' especifica marcadores de círculo vermelho ('o'), linha tracejada ('--') e cor vermelha ('r').

````python
import matplotlib.pyplot as plt

x = [4, 5, 6, 7, 8]
y = [5, 7, 4, 8, 6]

plt.plot(x, y, 'ro--', label='Círculos Vermelhos Tracejados')
plt.legend()
plt.show()
````

### 3.1.2 Gráficos de Dispersão

#### Os gráficos de dispersão são empregados para visualizar a relação entre duas variáveis numéricas, exibindo pontos de dados individuais no plano cartesiano . Eles são essenciais para identificar correlações e padrões em dados pareados, permitindo observar se existe alguma tendência ou agrupamento entre as duas variáveis analisadas.  

#### A criação de um gráfico de dispersão básico é feita com a função plt.scatter().

````python
import matplotlib.pyplot as plt
import numpy as np

# Gera dados simulados de idade e peso de atletas olímpicos
np.random.seed(42)  # Para reproduzir os mesmos resultados
idades = np.random.normal(25, 4, 200)
pesos = np.random.normal(75, 15, 200)
pesos += (idades - 25) * 1.5

# Gera dados de gênero aleatórios (50% homens, 50% mulheres)
generos = np.random.choice(['Homem', 'Mulher'], size=200)

# Separa os dados por gênero
idades_homens = idades[generos == 'Homem']
pesos_homens = pesos[generos == 'Homem']
idades_mulheres = idades[generos == 'Mulher']
pesos_mulheres = pesos[generos == 'Mulher']

# Cria o gráfico de dispersão com cores diferentes para cada gênero
plt.scatter(idades_homens, pesos_homens, label='Homens', alpha=0.7, color='blue')
plt.scatter(idades_mulheres, pesos_mulheres, label='Mulheres', alpha=0.7, color='red')

# Adiciona rótulos aos eixos e um título
plt.xlabel('Idade (anos)')
plt.ylabel('Peso (kg)')
plt.title('Idade vs. Peso de Atletas Olímpicos (por Gênero)')

# Adiciona legenda
plt.legend()

# Exibe o gráfico
plt.show()
````

#### A função plt.scatter() oferece diversas opções para personalizar a aparência dos marcadores:

#### **Tamanho**: O parâmetro s controla o tamanho dos marcadores. O valor de s é numérico e representa a área do marcador em pontos ao quadrado, por padrão. Isso significa que um valor maior de s resultará em marcadores maiores. Variar o tamanho dos marcadores pode representar uma terceira dimensão dos dados ou enfatizar certos pontos de acordo com sua importância ou valor.   

#### Por exemplo, se houver dados onde o tamanho do marcador deve refletir a magnitude de uma terceira variável, é possível passar uma lista ou array de tamanhos correspondente a cada ponto para o parâmetro s.

````python
import matplotlib.pyplot as plt

# Dados de amostra para as coordenadas x e y
x = [3, 4, 5, 6, 7]
y = [4, 6, 3, 7, 5]

# Dados de amostra para o tamanho dos marcadores
sizes = [100, 200, 300, 400, 800]  # Tamanhos dos marcadores

# Cria o gráfico de dispersão, definindo o tamanho dos marcadores
plt.scatter(x, y, s=sizes)

# Adiciona rótulos aos eixos
plt.xlabel('Eixo X')
plt.ylabel('Eixo Y')
plt.title('Gráfico de Dispersão com Tamanhos Variáveis')

# Exibe o gráfico
plt.show()
````

````python
import matplotlib.pyplot as plt

# Dados de amostra para as coordenadas x e y
x = [3, 4, 5, 6, 7]
y = [4, 6, 3, 7, 5]

# Define um tamanho único para todos os marcadores
marker_size = 100

# Cria o gráfico de dispersão com marcadores de tamanho uniforme
plt.scatter(x, y, s=marker_size)

# Adiciona rótulos aos eixos
plt.xlabel('Eixo X')
plt.ylabel('Eixo Y')
plt.title('Gráfico de Dispersão com Tamanho de Marcador Uniforme')

# Exibe o gráfico
plt.show()
````

#### **Cor**: É possível trocar a cor dos marcadores em gráficos de dispersão criados com a função plt.scatter() utilizando o parâmetro c.

#### O parâmetro c aceita diferentes tipos de argumentos para definir a cor dos marcadores:

- Uma única cor: especifica uma única cor para todos os marcadores usando um nome de cor (por exemplo, 'red', 'green', 'blue'), um código hexadecimal (por exemplo, '#FF0000'), ou uma tupla RGB ou RGBA (por exemplo, (1, 0, 0) para vermelho).
- Uma lista ou array de cores: Cada marcador tenha uma cor diferente, sendo que o comprimento dessa lista deve ser o mesmo que o número de pontos de dados.
- Um mapa de cores (colormap): é possível usar um mapa de cores para atribuir cores aos marcadores com base em um terceiro conjunto de dados. Para isso, é necessário passar os dados para o parâmetro c e especificar o mapa de cores desejado com o parâmetro cmap.

````python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [2, 4, 1, 5, 3]

plt.scatter(x, y, c='red')  # Define a cor de todos os marcadores para azul
plt.xlabel('Eixo X')
plt.ylabel('Eixo Y')
plt.title('Gráfico de Dispersão com Marcadores Vermelhos')
plt.show()
````

````python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [2, 4, 1, 5, 3]
colors = ['red', 'green', 'blue', 'yellow', 'purple']

plt.scatter(x, y, c=colors)  # Define uma cor diferente para cada marcador
plt.xlabel('Eixo X')
plt.ylabel('Eixo Y')
plt.title('Gráfico de Dispersão com Cores Variadas')
plt.show()
````

````python
import matplotlib.pyplot as plt
import numpy as np

x = np.random.rand(50)
y = np.random.rand(50)
z = np.random.rand(50)  # Uma terceira variável para definir as cores

plt.scatter(x, y, c=z, cmap='viridis')  # Usa o mapa de cores 'viridis' com base nos valores de z
plt.colorbar(label='Valor de Z')  # Adiciona uma barra de cores para referência
plt.xlabel('Eixo X')
plt.ylabel('Eixo Y')
plt.title('Gráfico de Dispersão com Cores Mapeadas')
plt.show()
````

### 3.1.3 Gráficos de Barras

#### Os gráficos de barras são utilizados para comparar dados categóricos, onde o comprimento ou a altura de cada barra representa o valor correspondente à categoria. 

#### Eles são eficazes para visualizar e comparar valores discretos entre diferentes categorias, tornando as comparações diretas e fáceis de interpretar.  

#### A criação de gráficos de barras verticais é feita utilizando a função plt.bar().

````python
import matplotlib.pyplot as plt

# Categorias e valores de amostra
categorias = ['A', 'B', 'C', 'D']  # Categorias para o gráfico
valores = [4, 6, 1, 7]  # Valores correspondentes às categorias

# Cria o gráfico de barras verticais
plt.bar(categorias, valores)

# Exibe o gráfico
plt.show()
````

#### Para criar gráficos de barras horizontais, utiliza-se a função plt.barh(), invertendo a orientação da representação visual. O código abaixo ilustra a criação de um gráfico de barras horizontal:

````python
import matplotlib.pyplot as plt

# Categorias e valores de amostra
categorias = ['A', 'B', 'C', 'D']  # Categorias para o gráfico
valores = [4, 6, 1, 7]  # Valores correspondentes às categorias

# Cria o gráfico de barras verticais
plt.barh(categorias, valores)

# Exibe o gráfico
plt.show()
````

#### Gráficos de barras agrupadas são utilizados para comparar múltiplos grupos de dados lado a lado, facilitando a análise das diferenças entre os grupos para cada categoria. A criação de gráficos de barras agrupadas envolve o ajuste da posição das barras no eixo para evitar sobreposição.

````python
import matplotlib.pyplot as plt
import numpy as np

# Categorias e valores de amostra para dois grupos
categorias = ['A', 'B', 'C', 'D']  # Categorias para o gráfico
valores1 = [4, 6, 1, 7]
valores2 = [8, 2, 5, 3]

# Define a posição das barras no eixo x
x = np.arange(len(categorias))
largura = 0.35  # Largura das barras

# Cria as barras para o primeiro grupo
plt.bar(x - largura/2, valores1, largura, label='Grupo 1')

# Cria as barras para o segundo grupo
plt.bar(x + largura/2, valores2, largura, label='Grupo 2')

# Define os rótulos do eixo x nas posições corretas
plt.xticks(x, categorias)

# Adiciona uma legenda
plt.legend()

# Exibe o gráfico
plt.show()
````

#### Gráficos de barras empilhadas são utilizados para mostrar a contribuição de múltiplas partes para um todo, onde as barras são empilhadas umas sobre as outras. A criação de gráficos de barras empilhadas é feita utilizando o parâmetro bottom na função plt.bar(), especificando a altura da barra inferior para que a próxima barra seja plotada acima dela.

````python
import matplotlib.pyplot as plt

# Categorias e valores de amostra para dois grupos
categorias = ['A', 'B', 'C', 'D']  # Categorias para o gráfico
valores1 = [4, 6, 1, 7]
valores2 = [8, 2, 5, 3]

# Cria as barras para o primeiro grupo
plt.bar(categorias, valores1, label='Grupo 1')

# Cria as barras para o segundo grupo, empilhadas sobre o primeiro
plt.bar(categorias, valores2, bottom=valores1, label='Grupo 2')

# Adiciona uma legenda
plt.legend()

# Exibe o gráfico
plt.show()
````

#### Os gráficos de barras oferecem flexibilidade na visualização de comparações categóricas, seja através de barras simples, comparações agrupadas ou mostrando relações parte-todo com o empilhamento.

| Tipo de Gráfico | Uso Principal | Função Matplotlib |
| :----------------- | :---------------------------------------------------------------------------- | :---------------- |
| Gráfico de Linhas | Mostrar tendências ao longo do tempo ou dados contínuos | `plt.plot()` |
| Gráfico de Dispersão | Visualizar a relação entre duas variáveis numéricas, mostrando pontos individuais | `plt.scatter()` |
| Gráfico de Barras | Comparar dados categóricos, com barras representando valores | `plt.bar()` / `plt.barh()` |

### 3.1.4 Histogramas:

#### Os histogramas são utilizados para exibir a distribuição de um único conjunto de dados numéricos, dividindo-o em intervalos (bins) e mostrando a frequência de valores em cada intervalo . Eles são cruciais para entender a frequência e a dispersão dos valores dos dados, permitindo identificar a forma da distribuição, como simetria, assimetria e a presença de valores atípicos.  

#### A criação de um histograma básico é feita utilizando a função plt.hist()

````python
import matplotlib.pyplot as plt
import numpy as np

# Dados de amostra (100 valores aleatórios)
data = np.random.randn(1000)

# Cria o histograma com 10 bins
plt.hist(data, bins=10)

# Exibe o gráfico
plt.show()
````

#### O número de intervalos (bins) em um histograma pode ser personalizado utilizando o parâmetro bins. O número de bins pode afetar significativamente a representação visual da distribuição; um número muito pequeno de bins pode ocultar detalhes importantes da distribuição, enquanto um número muito grande pode exibir ruído excessivo nos dados. A escolha apropriada do número de bins depende da natureza e do tamanho do conjunto de dados.

#### Para normalizar o histograma e exibir a densidade de probabilidade em vez da frequência bruta, utiliza-se o parâmetro density=True. Histogramas normalizados são particularmente úteis para comparar distribuições de conjuntos de dados com tamanhos de amostra diferentes, pois a área sob o histograma normalizado será igual a 1.

#### Os histogramas fornecem um resumo visual da distribuição de um conjunto de dados, permitindo a identificação rápida da tendência central, da dispersão e da assimetria dos dados, auxiliando na compreensão das características fundamentais da distribuição.

#### **Número de Bins**: o parâmetro bins define o número de barras (intervalos) no histograma. É possível especificar um número inteiro para a quantidade de bins ou uma sequência de valores que definem os limites de cada bin.

````python
import matplotlib.pyplot as plt
import numpy as np

data = np.random.randn(1000)

# Especifica um número diferente de bins
plt.hist(data, bins=30)
plt.title('Histograma com 30 Bins')
plt.show()

# Especifica os limites dos bins
bins_especificos = [-3, -2, -1, 0, 1, 2, 3]
plt.hist(data, bins=bins_especificos)
plt.title('Histograma com Bins Específicos')
plt.show()
````

#### **Cor das Barras**: é possível alterar a cor das barras usando o parâmetro color

````python
plt.hist(data, bins=20, color='skyblue')
plt.title('Histograma com Barras Azuis Claras')
plt.show()
````

#### **Cor das Bordas**: Para adicionar bordas às barras e definir sua cor, use os parâmetros edgecolor.

````python
plt.hist(data, bins=20, color='lightcoral', edgecolor='black')
plt.title('Histograma com Bordas Pretas')
plt.show()
````

#### **Estilo das Barras**: O parâmetro histtype permite escolher diferentes estilos para o histograma, como 'bar' (padrão), 'barstacked', 'step' (apenas a linha do contorno) e 'stepfilled' (contorno preenchido).

````python
plt.hist(data, bins=20, histtype='step')
plt.title('Histograma com Estilo "Step"')
plt.show()

plt.hist(data, bins=20, histtype='stepfilled', alpha=0.7) # alpha para transparência
plt.title('Histograma com Estilo "Stepfilled"')
plt.show()
````

#### **Alinhamento das Barras**: O parâmetro align controla como as barras são alinhadas em relação aos bins. As opções são 'left', 'mid' (padrão) e 'right'.

````python
plt.hist(data, bins=20, align='left')
plt.title('Histograma com Alinhamento à Esquerda')
plt.show()
````

#### **Adicionando Rótulos e Título**: Como em outros gráficos, é possível usar plt.xlabel(), plt.ylabel() e plt.title() para adicionar informações contextuais ao seu histograma.

````python
plt.hist(data, bins=25, color='gold', edgecolor='darkgoldenrod')
plt.xlabel('Valores')
plt.ylabel('Frequência')
plt.title('Distribuição dos Dados')
plt.grid(axis='y', alpha=0.5) # Adiciona linhas de grade no eixo y
plt.show()
````

#### **Histogramas Múltiplos**: é possível plotar múltiplos histogramas no mesmo gráfico para comparar distribuições. Basta chamar a função plt.hist() várias vezes com diferentes conjuntos de dados. O parâmetro label e a função plt.legend() são utilizados para identificar cada histograma.

````python
data1 = np.random.normal(0, 1, 1000)
data2 = np.random.normal(2, 1.5, 1000)

plt.hist(data1, bins=20, color='skyblue', alpha=0.5, label='Dados 1')
plt.hist(data2, bins=20, color='salmon', alpha=0.5, label='Dados 2')
plt.xlabel('Valores')
plt.ylabel('Frequência')
plt.title('Comparação de Distribuições')
plt.legend()
plt.show()
````

### 3.1.5 Gráficos de Pizza:

#### Os gráficos de pizza são gráficos circulares utilizados para representar proporções de um todo, onde cada fatia do círculo corresponde a uma categoria e o tamanho da fatia é proporcional à sua porcentagem do total . Eles são mais eficazes para mostrar a relação das partes com o todo, mas devem ser usados com cautela, pois podem ser menos precisos do que outros tipos de gráficos para comparar categorias de tamanhos semelhantes.  

#### A criação de um gráfico de pizza básico é feita utilizando a função plt.pie(). 

````python
import matplotlib.pyplot as plt

# Tamanhos das fatias e rótulos
tamanhos = [30, 20, 25, 25]  # Percentuais ou valores relativos
rotulos = ['A', 'B', 'C', 'D']  # Rótulos das fatias

# Cria o gráfico de pizza
plt.pie(tamanhos, labels=rotulos, autopct='%1.1f%%')  # Exibe percentuais

# Exibe o gráfico
plt.show()
````

#### **Alterar as Cores das Fatias**: é possível personalizar as cores das fatias do gráfico de pizza para torná-lo mais visualmente atraente ou para destacar categorias específicas.

````python
import matplotlib.pyplot as plt

tamanhos = [30, 20, 25, 25]
rotulos = ['A', 'B', 'C', 'D']
cores = ['gold', 'lightskyblue', 'lightcoral', 'lightgreen']  # Lista de cores

plt.pie(tamanhos, labels=rotulos, autopct='%1.1f%%', colors=cores)
plt.show()
````

### 3.1.6 Gráficos de Caixa:

#### Os gráficos de caixa (ou box-and-whisker plots) fornecem uma maneira padronizada de exibir a distribuição de dados com base em um resumo de cinco números: o mínimo, o primeiro quartil (Q1), a mediana (Q2), o terceiro quartil (Q3) e o máximo 1. Eles também podem indicar a presença de outliers, que são valores que se desviam significativamente do restante dos dados. 

#### Os gráficos de caixa são excelentes para comparar a distribuição de múltiplos conjuntos de dados e identificar características como assimetria e a presença de valores atípicos.

#### A criação de um gráfico de caixa básico é feita utilizando a função plt.boxplot().

````python
import matplotlib.pyplot as plt
import numpy as np

# Dados de amostra (múltiplos conjuntos de dados podem ser fornecidos como uma lista de arrays)
data = [np.random.normal(0, 1, 100), np.random.normal(2, 1.5, 150), np.random.normal(-1, 0.8, 120)]

# Cria o gráfico de caixa
plt.boxplot(data)

# Adiciona rótulos para os diferentes conjuntos de dados (opcional)
plt.xticks([1, 2, 3], ['Conjunto 1', 'Conjunto 2', 'Conjunto 3'])

# Exibe o gráfico
plt.show()
````

#### **cor e o estilo das caixas e bigodes**: A aparência das caixas e bigodes pode ser personalizada. O uso de patch_artist=True permite o preenchimento das caixas com cores, e a iteração sobre as caixas possibilita a definição de cores distintas. A cor e a espessura da linha da mediana também podem ser modificadas para aumentar a visibilidade.

````python
import matplotlib.pyplot as plt
import numpy as np

data = [np.random.normal(0, 1, 100), np.random.normal(2, 1.5, 150), np.random.normal(-1, 0.8, 120)]

# Customização das caixas e bigodes
boxplot = plt.boxplot(data, patch_artist=True)  # patch_artist=True permite preencher as caixas

colors = ['lightblue', 'lightgreen', 'lightcoral']
for patch, color in zip(boxplot['boxes'], colors):
    patch.set_facecolor(color)  # Cor de preenchimento das caixas

for median in boxplot['medians']:
    median.set(color='black', linewidth=2)  # Cor e espessura da linha da mediana

plt.xticks([1, 2, 3], ['Conjunto 1', 'Conjunto 2', 'Conjunto 3'])
plt.show()
````

#### **Adição de outliers personalizados**: outliers personalizados podem ser adicionados ao gráfico. Uma lista de outliers é criada para cada conjunto de dados e adicionada aos outliers existentes plotados pelo boxplot. Isso possibilita o destaque de pontos de dados específicos.

````python
import matplotlib.pyplot as plt
import numpy as np

# Gerando os dados principais
data = [np.random.normal(0, 1, 100), np.random.normal(2, 1.5, 150), np.random.normal(-1, 0.8, 120)]

# Adicionando outliers personalizados
outliers = [[-3, 3.5], [5], [-3]]  # Lista de outliers para cada conjunto de dados

# Criando o boxplot
boxplot = plt.boxplot(data, flierprops={'marker': 'o', 'markerfacecolor': 'red'})

# Adicionando os outliers manualmente com scatter
for i, outlier_values in enumerate(outliers):
    x_positions = np.full(len(outlier_values), i + 1)  # Posiciona no eixo X corretamente
    plt.scatter(x_positions, outlier_values, color='red', marker='o')

# Ajustando os rótulos do eixo X
plt.xticks([1, 2, 3], ['Conjunto 1', 'Conjunto 2', 'Conjunto 3'])

# Exibir o gráfico
plt.show()
````

#### **Alteração da orientação do gráfico de caixa**: a orientação do gráfico de caixa pode ser alterada de vertical para horizontal. A definição de vert=False no boxplot inverte a orientação dos eixos, o que pode ser útil para comparar conjuntos de dados lado a lado.

````python
import matplotlib.pyplot as plt
import numpy as np

data = [np.random.normal(0, 1, 100), np.random.normal(2, 1.5, 150), np.random.normal(-1, 0.8, 120)]

# Gráfico de caixa horizontal
plt.boxplot(data, vert=False)

plt.yticks([1, 2, 3], ['Conjunto 1', 'Conjunto 2', 'Conjunto 3'])
plt.show()
````

#### **Adição de título e rótulos aos eixos**: um título e rótulos podem ser adicionados aos eixos do gráfico. Isso melhora a clareza e a legibilidade do gráfico.

````python
import matplotlib.pyplot as plt
import numpy as np

data = [np.random.normal(0, 1, 100), np.random.normal(2, 1.5, 150), np.random.normal(-1, 0.8, 120)]

plt.boxplot(data)

plt.xticks([1, 2, 3], ['Conjunto 1', 'Conjunto 2', 'Conjunto 3'])
plt.title('Distribuição dos Dados')
plt.xlabel('Conjuntos de Dados')
plt.ylabel('Valores')
plt.show()
````

### 3.1.7 Gráficos de Violino:

#### O gráfico de violino é uma ferramenta de visualização de dados que combina características de um box plot e de um histograma.

#### Mostra a densidade de probabilidade dos dados em diferentes valores, permitindo identificar a forma da distribuição (simétrica, assimétrica, multimodal, etc.), facilita a comparação da forma, da amplitude e da posição das distribuições de dados entre diferentes grupos ou categorias.

#### A presença de outliers pode ser observada pela extensão dos "violinos" e pela posição dos marcadores de mediana e quartis.

#### O gráfico de violino é especialmente útil para visualizar dados com múltiplas modalidades (picos na distribuição), que podem não ser evidentes em um box plot.

#### A criação de um gráfico de violino básico é feita utilizando a função plt.violinplot().

````python
import matplotlib.pyplot as plt
import numpy as np

# Dados de exemplo
data = [np.random.normal(0, std, 100) for std in range(1, 4)]

# Criar o gráfico de violino
plt.violinplot(data, showmeans=False, showmedians=True)

# Personalização (opcional)
plt.xticks([1, 2, 3], ['Grupo 1', 'Grupo 2', 'Grupo 3'])
plt.xlabel('Grupos')
plt.ylabel('Valores')
plt.title('Gráfico de Violino de Exemplo')

# Exibir o gráfico
plt.show()
````

#### **Alterar a Cor dos Violinos**: é possível personalizar as cores dos violinos para destacar diferentes grupos ou para melhorar a estética visual do gráfico.

````python
import matplotlib.pyplot as plt
import numpy as np

data = [np.random.normal(0, std, 100) for std in range(1, 4)]

parts = plt.violinplot(data, showmeans=False, showmedians=True)

colors = ['lightblue', 'lightgreen', 'lightcoral']
for pc, color in zip(parts['bodies'], colors):
    pc.set_facecolor(color)

plt.xticks([1, 2, 3], ['Grupo 1', 'Grupo 2', 'Grupo 3'])
plt.xlabel('Grupos')
plt.ylabel('Valores')
plt.title('Gráfico de Violino Personalizado')
plt.show()
````

#### **Mostrar a Média e os Quartis**: além da mediana, é possível exibir a média e os quartis nos gráficos de violino para fornecer informações estatísticas adicionais sobre a distribuição dos dados.

````python
import matplotlib.pyplot as plt
import numpy as np

data = [np.random.normal(0, std, 100) for std in range(1, 4)]

plt.violinplot(data, showmeans=True, showextrema=True, quantiles=[[0.25, 0.75]] * 3)

plt.xticks([1, 2, 3], ['Grupo 1', 'Grupo 2', 'Grupo 3'])
plt.xlabel('Grupos')
plt.ylabel('Valores')
plt.title('Gráfico de Violino com Média e Quartis')
plt.show()
````

#### **Adicionar Marcadores Individuais de Dados**: para visualizar a distribuição de dados individuais dentro de cada violino, é possível adicionar marcadores de dados dispersos.

### 3.1.8 Gráficos de Contorno:

#### Os gráficos de contorno representam dados tridimensionais em duas dimensões utilizando linhas de contorno que conectam pontos de igual valor. 

#### Eles são úteis para visualizar superfícies, elevações ou qualquer quantidade tridimensional onde a terceira dimensão pode ser representada por níveis ou linhas. Os gráficos de contorno são amplamente utilizados em campos como geografia (mapas de elevação) e meteorologia (padrões climáticos).  

#### A criação de um gráfico de contorno básico é feita utilizando a função plt.contour(), e a criação de um gráfico de contorno preenchido é feita com plt.contourf(). 

#### Para utilizar essas funções, é necessário ter dados bidimensionais, que podem ser criados utilizando a função np.meshgrid() para gerar uma grade de coordenadas a partir de vetores unidimensionais.

````python
import matplotlib.pyplot as plt
import numpy as np

# Cria uma grade de coordenadas
x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)
X, Y = np.meshgrid(x, y)

# Define uma função para a terceira dimensão (Z)
Z = np.sin(np.sqrt(X**2 + Y**2))

# Cria o gráfico de contorno preenchido
plt.contourf(X, Y, Z, cmap='viridis')

# Adiciona uma barra de cores para interpretar os níveis
plt.colorbar()

# Exibe o gráfico
plt.show()
````

#### **Alterar o Número de Níveis de Contorno**: o número de níveis de contorno determina a granularidade da representação da função. Ajustar esse número pode revelar detalhes finos ou fornecer uma visão geral mais suave da distribuição dos dados.

````python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)
X, Y = np.meshgrid(x, y)
Z = np.sin(np.sqrt(X**2 + Y**2))

plt.contourf(X, Y, Z, levels=20, cmap='viridis')  # 20 níveis de contorno
plt.colorbar()
plt.show()
````

#### **Adicionar Linhas de Contorno**: além do preenchimento de contorno, é possível adicionar linhas de contorno para destacar as fronteiras entre os níveis. Isso pode melhorar a clareza do gráfico, especialmente em áreas de transição rápida.

````python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)
X, Y = np.meshgrid(x, y)
Z = np.sin(np.sqrt(X**2 + Y**2))

plt.contourf(X, Y, Z, cmap='viridis')
plt.contour(X, Y, Z, colors='black', linewidths=0.5)  # Adiciona linhas de contorno pretas
plt.colorbar()
plt.show()
````

#### **Alterar o Mapa de Cores (Colormap)**: o mapa de cores define como os valores da terceira dimensão (Z) são mapeados para cores. A escolha de um mapa de cores adequado pode realçar padrões específicos nos dados e melhorar a interpretação visual.

````python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)
X, Y = np.meshgrid(x, y)
Z = np.sin(np.sqrt(X**2 + Y**2))

plt.contourf(X, Y, Z, cmap='RdBu_r')  # Mapa de cores vermelho-azul invertido
plt.colorbar()
plt.show()
````

### 3.1.9 Gráfico de Fluxo:

#### O gráfico de fluxo, ou stream plot, é uma ferramenta de visualização de dados que representa o movimento de um campo vetorial. 

#### Sua utilidade reside na capacidade de visualizar trajetórias, as linhas de fluxo indicam o caminho que partículas seguiriam ao se moverem através do campo vetorial, permitindo a compreensão de padrões de movimento.

#### O gráfico revela a direção e a intensidade do campo vetorial em diferentes pontos, facilitando a identificação de áreas de alta e baixa velocidade, bem como de pontos de convergência e divergência

#### Em sistemas com fluxos complexos, como fluidos, correntes de ar ou campos magnéticos, o gráfico de fluxo auxilia na identificação de padrões que seriam difíceis de discernir de outra forma alem de aplicações em diversas áreas, incluindo meteorologia (visualização de padrões de vento), engenharia (análise de fluxo de fluidos), física (estudo de campos magnéticos) e ciência da computação (visualização de algoritmos de otimização).


````python
import numpy as np
import matplotlib.pyplot as plt

# 1. Criar uma grade de pontos (x, y)
largura, altura = 10, 8
x, y = np.meshgrid(np.linspace(0, largura, 20), np.linspace(0, altura, 20))

# 2. Definir o campo vetorial (u, v)
# Neste exemplo, o campo vetorial simula um fluxo rotacional
u = -y + altura / 2
v = x - largura / 2

# 3. Criar o gráfico de fluxo
plt.figure(figsize=(8, 6))
plt.streamplot(x, y, u, v, density=1.2, color=np.sqrt(u**2 + v**2), cmap='viridis')

# 4. Adicionar detalhes ao gráfico
plt.title('Gráfico de Fluxo Rotacional')
plt.xlabel('Eixo X')
plt.ylabel('Eixo Y')
plt.colorbar(label='Velocidade')  # Adiciona uma barra de cores para a velocidade
plt.grid(True)
plt.show()
````

#### **Alterar a Densidade das Linhas de Fluxo**: a densidade das linhas de fluxo controla o número de linhas exibidas no gráfico. Ajustar a densidade pode tornar o gráfico mais detalhado ou mais simplificado, dependendo da necessidade de visualização.

````python
import numpy as np
import matplotlib.pyplot as plt

largura, altura = 10, 8
x, y = np.meshgrid(np.linspace(0, largura, 20), np.linspace(0, altura, 20))
u = -y + altura / 2
v = x - largura / 2

plt.figure(figsize=(8, 6))
plt.streamplot(x, y, u, v, density=2, color=np.sqrt(u**2 + v**2), cmap='viridis')  # Densidade aumentada
plt.title('Gráfico de Fluxo com Densidade Ajustada')
plt.xlabel('Eixo X')
plt.ylabel('Eixo Y')
plt.colorbar(label='Velocidade')
plt.grid(True)
plt.show()
````

### 3.1.10 Gráfico de Mapa de Calor:

#### O mapa de calor é usado para visualizar dados matriciais, onde as cores representam os valores dos elementos da matriz.

### 3.1.11 Gráfico de Rede:

#### O gráfico de rede é usado para visualizar relações entre entidades, onde os nós representam as entidades e as arestas representam as relações.

#### Ele é útil para analisar estruturas complexas, como redes sociais, redes de computadores e redes de proteínas.

#### Dentre suas utilizações estão a visualização as conexões entre amigos em uma rede social, analisar a estrutura de uma rede de computadores, mostrar as interações entre proteínas em uma célula, visualizar o fluxo de passageiros entre aeroportos.

````python
import matplotlib.pyplot as plt

# 1. Dados da rede (aeroportos e conexões)
aeroportos = {
    'São Paulo': (23.5505, 46.6333),  # Latitude e longitude
    'Rio de Janeiro': (22.9068, 43.1729),
    'Brasília': (15.7939, 47.8828),
    'Belo Horizonte': (19.9167, 43.9345),
    'Porto Alegre': (30.0346, 51.2065)
}

conexoes = [
    ('São Paulo', 'Rio de Janeiro'),
    ('São Paulo', 'Brasília'),
    ('São Paulo', 'Belo Horizonte'),
    ('São Paulo', 'Porto Alegre'),
    ('Rio de Janeiro', 'Brasília'),
    ('Belo Horizonte', 'Brasília'),
    ('Porto Alegre', 'Belo Horizonte')
]

# 2. Plotagem dos aeroportos
x_aeroportos = [aeroportos[cidade][1] for cidade in aeroportos]  # Longitude (x)
y_aeroportos = [aeroportos[cidade][0] for cidade in aeroportos]  # Latitude (y)
plt.scatter(x_aeroportos, y_aeroportos, s=500, c='skyblue')

# 3. Plotagem das conexões
for origem, destino in conexoes:
    x_conexao = [aeroportos[origem][1], aeroportos[destino][1]]
    y_conexao = [aeroportos[origem][0], aeroportos[destino][0]]
    plt.plot(x_conexao, y_conexao, c='gray')

# 4. Rótulos dos aeroportos
for cidade, (lat, lon) in aeroportos.items():
    plt.annotate(cidade, xy=(lon, lat), ha='center', va='center', fontsize=10)

# 5. Detalhes do gráfico
plt.title('Conexões Aéreas entre Cidades')
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.grid(True)
plt.show()
````

````python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import numpy as np

# Cria dados de amostra 3D
x = np.random.normal(0, 1, 100)
y = np.random.normal(0, 1, 100)
z = np.random.normal(0, 1, 100)

# Cria uma figura e adiciona um subplot 3D
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

# Cria o gráfico de dispersão 3D
ax.scatter(x, y, z)

# Define os rótulos dos eixos
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')

# Exibe o gráfico
plt.show()
````

#### Um gráfico de superfície 3D pode ser criado utilizando a função ax.plot_surface(), que requer dados bidimensionais para representar a superfície.

## 3.2 Gráficos com Object-oriented API

#### A API Orientada a Objetos (OO) do Matplotlib é uma forma de interagir com a biblioteca de visualização de dados, permitindo um controle mais granular e flexível sobre a criação de gráficos. Em vez de usar funções globais que modificam um estado implícito, a API OO trabalha com objetos que representam os elementos do gráfico, como figuras, eixos e artistas.

#### Conceitos-chave:
- Abstração: A API OO abstrai os elementos de um gráfico em objetos, como Figure (figura) e Axes (eixos). Isso permite que você trate cada elemento como uma entidade independente, com suas próprias propriedades e métodos.
- Encapsulamento: Cada objeto encapsula seus dados e comportamentos. Por exemplo, o objeto Axes encapsula os dados a serem plotados, os rótulos dos eixos e os métodos para desenhar os dados.
- Hierarquia: Os objetos são organizados em uma hierarquia. A figura é o contêiner principal, que contém os eixos, que por sua vez contêm os artistas (elementos visuais). Essa hierarquia permite um controle organizado sobre os elementos do gráfico.
- Flexibilidade: A API OO oferece uma grande flexibilidade na personalização de gráficos. Você pode acessar e modificar as propriedades de cada objeto, permitindo um controle fino sobre a aparência do gráfico.
- Reusabilidade: A abordagem orientada a objetos facilita a criação de código reutilizável. Você pode criar funções ou classes que geram gráficos personalizados, que podem ser reutilizados em diferentes partes do seu código.

#### **Figura** (Figure): O Contêiner Principal

#### Definição:
- A Figura é o objeto de nível superior que representa toda a área de desenho. É como a "tela" onde todos os elementos do gráfico são colocados.
- Pode conter um ou mais objetos Axes, além de outros elementos como títulos gerais e legendas.

#### Características Principais:
- Contêiner: Atua como o contêiner para todos os outros elementos do gráfico.
- Múltiplas Figuras: É possível criar várias Figuras em um mesmo script, permitindo a geração de múltiplos gráficos independentes.
- Controle Geral: Permite controlar propriedades gerais do gráfico, como tamanho, resolução e cor de fundo.

#### Métodos Importantes:

```
        fig.add_subplot(): Adiciona um novo subplot (eixos) à figura.
        fig.suptitle(): Adiciona um título geral à figura.
        fig.savefig(): Salva a figura em um arquivo de imagem.
```

#### Exemplo:

1. Cria-se um objeto Figure com um tamanho específico.
2. Adiciona-se um objeto Axes à figura.
3. Adiciona-se um título geral à figura.
4. Plotam-se dados simples nos eixos.
5. Salva-se a figura em um arquivo.

````python
import matplotlib.pyplot as plt

# Cria uma figura com tamanho personalizado
fig = plt.figure(figsize=(8, 6))

# Adiciona um subplot (eixos) à figura
ax = fig.add_subplot(111)  # 111 significa 1 linha, 1 coluna, 1º gráfico

# Adiciona um título geral à figura
fig.suptitle('Exemplo de Figura')

# Plota alguns dados simples
ax.plot([1, 2, 3, 4], [10, 15, 13, 18])

# Salva a figura em um arquivo
fig.savefig('exemplo_figura.png')

# Mostra a figura
plt.show()
````

#### **Eixos** (Axes): A Área de Plotagem

#### Definição:
- Os Eixos são a área onde os dados são realmente plotados.
- Cada Figura pode conter um ou mais objetos Axes, organizados em subplots.
- Os Eixos definem os limites dos dados, os rótulos dos eixos e outros elementos visuais específicos da área de plotagem.
        
#### Características Principais:
- Área de Plotagem: Representa a área retangular onde os dados são visualizados.
- Controle dos Dados: Define os limites dos dados nos eixos x e y.
- Personalização Específica: Permite personalizar rótulos, títulos, grades e outros elementos específicos da área de plotagem.
        
#### Métodos Importantes:

```
        ax.plot(): Plota linhas.
        ax.scatter(): Plota pontos.
        ax.bar(): Plota gráficos de barras.
        ax.set_xlabel(), ax.set_ylabel(): Define os rótulos dos eixos.
        ax.set_title(): Define o título dos eixos.
        ax.legend(): Adiciona uma legenda.
        ax.grid(): Adiciona uma grade.
```

#### Exemplo:
1. Cria-se um objeto Axes.
2. Plotam-se dados nos eixos, personalizando a cor, o estilo da linha e a largura.
3. Definem-se os rótulos dos eixos e o título.
4. Adiciona-se uma legenda e grade.

````python
import matplotlib.pyplot as plt
import numpy as np

# Cria-se uma figura e eixos
fig, ax = plt.subplots()

# Dados de exemplo
x = np.linspace(0, 10, 100)
y = np.sin(x)

# Plota-se os dados nos eixos
ax.plot(x, y, color='green', linestyle='--', linewidth=2, label='sen(x)')

# Define-se os rótulos dos eixos e o título
ax.set_xlabel('Eixo X')
ax.set_ylabel('Eixo Y')
ax.set_title('Gráfico de Seno')

# Adiciona-se uma legenda e grade
ax.legend()
ax.grid(True)

# Exibe-se o gráfico
plt.show()
````

#### **Artistas** (Artists): Os Elementos Visuais

#### Definição:
- Artistas são todos os elementos visuais que aparecem em uma Figura, como linhas, pontos, texto, legendas, etc.
- Os Eixos contêm muitos Artistas, e você pode personalizá-los para controlar a aparência do seu gráfico.
- 
#### Características Principais:
- Elementos Gráficos: Representam todos os elementos que podem ser desenhados em uma Figura.
- Personalização Detalhada: Permitem controlar propriedades específicas de cada elemento, como cor, estilo de linha, tamanho da fonte, etc.
- Hierarquia: Os Artistas são organizados em uma hierarquia dentro dos Eixos e da Figura.
- 
#### Exemplos de Artistas:
- Linhas (Line2D)
- Pontos (Scatter)
- Textos (Text)
- Legendas (Legend)
- Retângulos (Rectangle)

#### Exemplos:
1. Um artista de texto.
2. Um artista de retângulo.
3. Um artista de seta.

````python
import matplotlib.pyplot as plt

# Cria-se uma figura e eixos
fig, ax = plt.subplots()

# Adiciona-se um artista de texto
text = ax.text(0.5, 0.5, 'Texto de Exemplo', fontsize=16, color='red', ha='center', va='center')

# Adiciona-se um artista de retângulo
rect = plt.Rectangle((0.1, 0.1), 0.3, 0.4, facecolor='yellow', alpha=0.5)
ax.add_patch(rect)

# Adiciona-se um artista de seta
arrow = ax.annotate('Aponta aqui', xy=(0.8, 0.8), xytext=(0.6, 0.7), arrowprops=dict(facecolor='blue'))

# Define-se os limites dos eixos
ax.set_xlim(0, 1)
ax.set_ylim(0, 1)

# Exibe-se o gráfico
plt.show()
````

### 3.2.4 Trabalhando com Múltiplos Subplots (API Orientada a Objetos):

#### A API Orientada a Objetos é particularmente adequada para a criação de figuras contendo múltiplos subplots, permitindo a comparação de diferentes conjuntos de dados ou a visualização de diferentes aspectos dos mesmos dados. A função plt.subplots(nrows, ncols) simplifica a criação de uma grade de subplots, retornando um objeto Figure e um array de objetos Axes.

#### Neste código, plt.subplots(2, figsize=(7,4)) cria uma Figure e um array axs contendo dois objetos Axes, um para cada subplot. 

#### Cada subplot é acessado através da indexação do array axs (por exemplo, axs para o primeiro subplot e axs para o segundo), e os métodos de plotagem e personalização são chamados em cada objeto Axes individualmente. Isso permite que cada subplot seja configurado de forma independente, com seus próprios dados, título e rótulos. 

#### A função plt.tight_layout() ajusta automaticamente o espaçamento entre os subplots para evitar sobreposições. Para layouts de subplots mais complexos, o Matplotlib oferece o módulo GridSpec, que proporciona ainda maior flexibilidade no posicionamento e dimensionamento dos subplots.

````python
import matplotlib.pyplot as plt
import numpy as np

# Dados de amostra
x = np.linspace(0, 2 * np.pi, 400)
y1 = np.sin(x**2)
y2 = np.cos(x**2)

# Cria subplots utilizando a interface orientada a objetos
fig, axs = plt.subplots(2, figsize=(7, 4))
fig.suptitle('Criando Subplots na interface explícita')

# Plota os dados no primeiro subplot
axs[0].plot(x, y1)  # Use axs[0] para acessar o primeiro subplot
axs[0].set_title('Gráfico de Seno')

# Plota os dados no segundo subplot
axs[1].plot(x, y2)  # Use axs[1] para acessar o segundo subplot
axs[1].set_title('Gráfico de Cosseno')

# Ajusta o layout para evitar sobreposição de títulos
plt.tight_layout()

# Exibe o gráfico
plt.show()
````

### 3.2.5 Salvando e Exportando Gráficos:

#### A função plt.savefig() é utilizada para salvar os gráficos criados em arquivos. 

#### Salvar os gráficos é essencial para compartilhar as visualizações e incorporá-las em relatórios ou apresentações. O nome do arquivo e o formato (por exemplo, 'meu_grafico.png', 'relatorio_figura.pdf', 'grafico_interativo.svg') podem ser especificados como argumentos para a função. 

#### Diferentes formatos de arquivo são adequados para diferentes propósitos; PNG é um formato raster ideal para imagens na web, enquanto PDF e SVG são formatos vetoriais mais adequados para impressão e escalabilidade. 

#### A função plt.savefig() oferece várias opções importantes para controlar a saída do arquivo:

- dpi: Controla a resolução da imagem em pontos por polegada (dots per inch). Um valor de DPI mais alto resulta em imagens de maior resolução, o que é importante para publicações impressas ou quando detalhes finos são necessários.   
- transparent=True: Define o fundo do gráfico como transparente. A transparência é útil quando o gráfico precisa ser sobreposto em um fundo colorido ou em outra imagem.  
- bbox_inches='tight': Remove o espaço em branco extra ao redor do gráfico, resultando em uma imagem mais recortada. Isso pode tornar os gráficos mais limpos e focados nos dados. 
- pad_inches: Controla a quantidade de preenchimento (padding) ao redor do gráfico quando bbox_inches='tight' é utilizado. Permite um ajuste fino do espaçamento ao redor da imagem do gráfico.

#### Outras formas de salvar gráficos incluem o uso de plt.imsave() para salvar arrays como arquivos de imagem.

## 3.3 API Orientada a Objetos vs. API Pyplot: Uma Comparação Detalhada

#### A escolha entre a API Orientada a Objetos e a API Pyplot depende do nível de controle e da complexidade do gráfico desejado.

- Gerenciamento de Figuras e Eixos:
  - A principal diferença reside no gerenciamento de Figure e Axes.
  - Na API OO, a criação e manipulação desses objetos é explícita, proporcionando controle direto sobre cada elemento do gráfico.
  - Já na API Pyplot, o gerenciamento é implícito, com a biblioteca cuidando da criação e do rastreamento da figura e dos eixos atuais.

- Sintaxe e Estrutura do Código:
  - Na API OO, os métodos de plotagem são chamados no objeto Axes (por exemplo, ax.plot()), enquanto na Pyplot, as funções são chamadas diretamente no módulo plt (por exemplo, plt.plot()).
  - A API OO promove uma estrutura de código mais organizada, especialmente para gráficos complexos com múltiplos elementos, pois mantém referências explícitas aos objetos que estão sendo manipulados.

- Nível de Controle e Flexibilidade:
  - A API OO oferece um grau significativamente maior de controle e flexibilidade sobre os elementos do gráfico e suas propriedades, pois permite a manipulação direta dos objetos.
  - A natureza implícita da Pyplot pode dificultar a realização de personalizações específicas ou o gerenciamento independente de múltiplos gráficos.

#### A tabela a seguir resume as principais diferenças entre as duas APIs:

| Característica   | API Orientada a Objetos | API Pyplot               |
|-------------------|-------------------------|--------------------------|
| Figure/Axes       | Criados e gerenciados explicitamente | Gerenciados implicitamente |
| Controle          | Alto, manipulação direta de objetos | Menor, baseado em estados |
| Sintaxe (plotagem)| `ax.plot(x, y)`        | `plt.plot(x, y)`         |
| Complexidade      | Recomendada para gráficos complexos   | Adequada para gráficos simples |
| Estrutura         | Mais estruturada, melhor para projetos grandes | Menos estruturada, boa para scripts rápidos |
| Flexibilidade     | Altamente flexível e personalizável | Menos flexível para necessidades avançadas |

## 4 Considerações finais

#### O Matplotlib é uma das bibliotecas mais versáteis para visualização de dados em Python, oferecendo duas interfaces principais para criação de gráficos: a API Pyplot e a API Orientada a Objetos (OO).

#### A API Pyplot funciona como um estado global semelhante ao MATLAB, simplificando a criação de gráficos ao gerenciar automaticamente as figuras e os eixos. Sua sintaxe concisa torna essa abordagem ideal para prototipagem rápida, exploração de dados e geração de visualizações simples.

#### Apesar da praticidade, a API Pyplot pode se tornar limitada quando é necessário um controle mais preciso sobre múltiplos subplots, estilos complexos ou personalizações detalhadas.

#### Já a API Orientada a Objetos permite uma abordagem mais estruturada, oferecendo controle explícito sobre cada elemento da visualização, como figuras (Figure), eixos (Axes) e anotações. Essa abordagem é recomendada quando se trabalha com gráficos interativos, painéis de múltiplas visualizações, personalizações avançadas e integrações em aplicações maiores.

#### Ao criar explicitamente os objetos Figure e Axes, essa abordagem permite manipular cada gráfico de forma modular e reutilizável, tornando-a ideal para dashboards interativos, publicações científicas e relatórios técnicos.

#### Outro benefício da API OO é sua compatibilidade com outras bibliotecas, como Seaborn e Pandas, permitindo a incorporação de gráficos mais elaborados sem perder a flexibilidade.

#### Qual abordagem escolher?
- Use Pyplot se estiver criando gráficos rápidos, explorando dados interativamente ou trabalhando com scripts curtos.
- Prefira a API OO quando precisar de personalizações detalhadas, múltiplos subplots, gráficos interativos ou integração com outras bibliotecas.
