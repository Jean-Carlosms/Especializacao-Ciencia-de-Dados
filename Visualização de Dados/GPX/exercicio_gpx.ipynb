#### No ciclismo, diversas métricas são usadas para monitorar o desempenho e otimizar os treinos.

#### Essas métricas ajudam os ciclistas a ajustar seus treinos, avaliar seu progresso e melhorar seu desempenho. As principais métricas incluem:
1.	**Velocidade**: Mede a rapidez com que o ciclista pedala, geralmente em quilômetros por hora (km/h) ou milhas por hora (mph). Pode ser média ou instantânea.
2.	**Distância**: A distância total percorrida durante o treino ou prova, geralmente medida em quilômetros ou milhas.
3.	**Cadência**: Refere-se ao número de rotações do pedivela por minuto (RPM). Uma cadência eficiente geralmente está entre 80-100 RPM, dependendo do tipo de treino e terreno.
4.	**Potência (Watts)**: Mede o esforço gerado pelos músculos. O medidor de potência é uma ferramenta fundamental no ciclismo de alto desempenho, permitindo medir a força exercida de forma precisa. A potência é comumente medida em watts, sendo que mais watts indicam mais esforço.
5.	**Frequência Cardíaca (FC)**: Mede a quantidade de batimentos do coração por minuto. A frequência cardíaca é importante para monitorar a intensidade do treino e garantir que o ciclista esteja trabalhando nas zonas de esforço desejadas (aeróbica, anaeróbica, etc.).
6.	**Variação da Potência Normalizada (NP)**: Um cálculo que ajusta a média de potência para refletir as flutuações durante o treino, ajudando a entender melhor a intensidade real do esforço.
7.	**Excedente de Potência**: Refere-se ao esforço acima do nível básico necessário para manter a velocidade de deslocamento.
8.	**Índice de Esforço (TSS - Training Stress Score)**: Quantifica a carga do treino com base na potência, levando em conta a intensidade e duração. O TSS ajuda a planejar a recuperação e a progressão do treino.
9.	**Índice de Performance (IF - Intensity Factor)**: É a razão entre a potência normalizada e a potência limiar funcional (FTP - Functional Threshold Power). O IF ajuda a determinar a intensidade de um treino em relação ao FTP do ciclista.
10.	**Limiar de Potência Funcional (FTP)**: O maior valor médio de potência que o ciclista consegue sustentar por uma hora. O FTP é usado como um parâmetro para definir zonas de treino específicas (zona de recuperação, zona aeróbica, etc.).
11.	**Variação de Frequência Cardíaca (HRV)**: Mede a variabilidade dos intervalos entre os batimentos cardíacos. Essa métrica pode indicar o nível de estresse do corpo e a recuperação.

#### O FTP (Functional Threshold Power) é uma métrica muito importante no ciclismo, especialmente para ciclistas que desejam monitorar e melhorar seu desempenho.

#### Ele representa a maior potência média que um ciclista pode manter de forma sustentável por uma hora. Em outras palavras, é o limite de potência que o ciclista consegue manter sem sofrer uma fadiga excessiva. O FTP é usado para definir zonas de treinamento e ajudar os atletas a treinar de forma mais eficiente, maximizando os ganhos de desempenho.

#### Como o FTP é Determinado:

O FTP pode ser determinado de várias maneiras, mas a forma mais comum é através de um teste específico. Existem basicamente dois tipos principais de testes:

1. Teste de 20 minutos:
- O ciclista pedala o mais forte possível durante 20 minutos, e a potência média durante esse período é medida.
- Para estimar o FTP, pega-se 95% da potência média obtida nos 20 minutos.
- Exemplo: Se você conseguir uma potência média de 300 watts durante o teste de 20 minutos, seu FTP seria 300 * 0,95 = 285 watts.

2. Teste de 60 minutos (menos comum, mas mais preciso):
- O ciclista pedala durante 60 minutos contínuos na maior potência possível que consegue sustentar sem entrar em fadiga extrema.
- Esse é o método “padrão” de medir o FTP, mas na prática, muitos ciclistas não fazem o teste completo de uma hora por ser muito extenuante. Por isso, o teste de 20 minutos é mais popular, com o cálculo de 95% da média de potência.

#### Por Que o FTP é Importante?

#### O FTP ajuda os ciclistas a entenderem qual é a sua capacidade de manter uma alta intensidade por um período prolongado. Ao medir o FTP, os ciclistas podem adaptar seus treinos e monitorar seu progresso. O FTP também é usado para:
- Definir zonas de treinamento: A partir do FTP, é possível criar zonas de potência para diferentes tipos de treinamento, como recuperação, endurance, tempo, intervalo e potência máxima.
- Ajustar planos de treinamento: Com o FTP, é possível ajustar a intensidade dos treinos, otimizando o rendimento e evitando o overtraining (treinamento excessivo).
- Monitoramento de desempenho: O FTP também é uma excelente ferramenta para ver a evolução do ciclista ao longo do tempo, ajudando a avaliar os resultados do treinamento.

#### Com o FTP, os ciclistas podem dividir sua potência em diferentes zonas de treinamento, cada uma com objetivos específicos. As zonas geralmente são as seguintes:
1.	**Zona 1 - Recuperação (0-55% do FTP)**: Intensidade muito baixa, usada para recuperação e passeios leves.
2.	**Zona 2 - Endurance (56-75% do FTP)**: Treinamento de resistência, usado para melhorar a capacidade aeróbica e resistência geral.
3.	**Zona 3 - Tempo (76-90% do FTP)**: Intensidade moderada, focada em melhorar a capacidade de manter uma potência alta por períodos prolongados. Essa zona é comum em treinos de contrarrelógio e provas longas.
4.	**Zona 4 - Limiar (91-105% do FTP)**: Treinamento de alta intensidade, visando melhorar a capacidade de sustentar o esforço próximo ao FTP, usado para aumentar o limiar de lactato (o ponto onde o corpo começa a acumular lactato mais rapidamente).
5.	**Zona 5 - VO2 Máximo (106-120% do FTP)**: Intensidade alta, focada no aumento do VO2 máximo, que é a quantidade de oxigênio que o corpo consegue utilizar durante o exercício. Usado em treinos intervalados.
6.	**Zona 6 - Anaerobic0 (121-150% do FTP)**: Intensidade muito alta, aumentar a potência máxima e a capacidade de fazer esforços explosivos, como sprints.
7.	**Zona 7 - Sprint (150%+ do FTP)**: Zona de máxima intensidade, usada para esforços de curta duração e máxima explosão, como sprints finais em uma corrida.

#### O formato GPX (GPS Exchange Format) é um formato de arquivo XML amplamente utilizado para intercâmbio de dados de GPS entre dispositivos e aplicativos.

#### Ele foi desenvolvido para armazenar informações de percurso, como coordenadas geográficas (latitude e longitude), rotas, pontos de interesse (POIs) e tracks (trilhas).

#### Esses arquivos são especialmente úteis para ciclistas, corredores, caminhantes, e outros usuários de dispositivos GPS e são importados pelas plataformas de análise como o Strava por exemplo.

```xml
...
   <trkpt lat="-23.5002710" lon="-47.4280250">
    <ele>569.0</ele>
    <time>2025-01-12T08:44:02Z</time>
    <extensions>
     <power>0</power>
     <gpxtpx:TrackPointExtension>
      <gpxtpx:atemp>24</gpxtpx:atemp>
      <gpxtpx:hr>112</gpxtpx:hr>
      <gpxtpx:cad>0</gpxtpx:cad>
     </gpxtpx:TrackPointExtension>
    </extensions>
   </trkpt>
   <trkpt lat="-23.5002430" lon="-47.4280280">
    <ele>568.8</ele>
    <time>2025-01-12T08:44:03Z</time>
    <extensions>
     <power>0</power>
     <gpxtpx:TrackPointExtension>
      <gpxtpx:atemp>24</gpxtpx:atemp>
      <gpxtpx:hr>112</gpxtpx:hr>
      <gpxtpx:cad>0</gpxtpx:cad>
     </gpxtpx:TrackPointExtension>
    </extensions>
   </trkpt>
...
```

- **lat e lon**: Coordenadas geográficas (latitude e longitude) do ponto.
- **\<ele\>**: Elevação (altitude) do ponto em metros (neste caso, 569 metros).
- **\<time\>**: Hora exata em que o ponto foi registrado.
- **\<extensions\>**: Elemento que pode incluir dados adicionais não padrão no GPX, como potência, temperatura, frequência cardíaca, cadência, entre outros.
- **\<power\>**: Potência em watts (neste caso, 37 watts).
- **\<gpxtpx:TrackPointExtension\>**: Namespace que define extensões adicionais específicas da Garmin (como temperatura, frequência cardíaca e cadência).
- **\<gpxtpx:atemp\>**: Temperatura ambiente em graus Celsius (neste caso, 24°C).
- **\<gpxtpx:hr\>**: Frequência cardíaca em batimentos por minuto (neste caso, 112 bpm).
- **\<gpxtpx:cad\>:** Cadência (número de rotações do pedal por minuto) - neste caso, 0, indicando que o sensor de cadência não foi ativado ou não estava disponível.

````python
#Bibliotecas
import pandas as pd
import matplotlib.pyplot as plt
import gpxpy
import gpxpy.gpx
from geopy.distance import geodesic
from matplotlib.lines import Line2D
import folium
````

````python
# Carregar o arquivo GPX
with open("/content/atividade_1.gpx", "r") as gpx_file:
    gpx = gpxpy.parse(gpx_file)

# Lista para armazenar os dados
data = []
prev_point = None  # Para armazenar o ponto anterior e calcular a distância
coordinates = []

# Iterar pelos pontos da trilha
for track in gpx.tracks:
    for segment in track.segments:
        for point in segment.points:
            lat = point.latitude
            lon = point.longitude
            ele = point.elevation
            time = point.time
            coordinates.append([point.latitude, point.longitude])
            power = None
            atemp = None
            hr = None
            cad = None

            # Verificar extensões
            for extension in point.extensions:
                if extension.tag == "{http://www.garmin.com/xmlschemas/TrackPointExtension/v1}TrackPointExtension":
                    for child in extension:
                        if "atemp" in child.tag:
                            atemp = float(child.text)
                        elif "hr" in child.tag:
                            hr = int(child.text)
                        elif "cad" in child.tag:
                            cad = int(child.text)
                elif "power" in extension.tag:
                    power = int(extension.text)

            # Calcular a distância do ponto anterior
            distancia = geodesic((prev_point.latitude, prev_point.longitude), (lat, lon)).meters if prev_point else 0
            prev_point = point  # Atualizar o ponto anterior

            # Adicionar ao dataset
            data.append([lat, lon, ele, time, power, atemp, hr, cad, distancia])
````

````python
# Criar DataFrame
df = pd.DataFrame(data, columns=["Latitude", "Longitude", "Altitude", "Tempo", "Potência", "Temperatura", "Freq_Cardiaca", "Cadência", "Distância_metros"])

# Adicionar distância acumulada
df["Distância_acumulada"] = df["Distância_metros"].cumsum()

df['diferenca_elevacao'] = df['Altitude'].diff()
````

````python
# Calcular frequência cardíaca máxima e média
fc_max = df["Freq_Cardiaca"].max()
fc_media = df["Freq_Cardiaca"].mean()
ele_max = df["Altitude"].max()
ele_min = df["Altitude"].min()
ele_total = df["Altitude"].max()-df["Altitude"].min()

````

````python
# Criar o gráfico
plt.figure(figsize=(10, 5))
plt.plot(df["Distância_acumulada"], df["Freq_Cardiaca"], label="Frequência Cardíaca", color="red", linewidth=2)

# Adicionar título e rótulos
plt.title("Frequência Cardíaca ao Longo da Distância Percorrida")
plt.xlabel("Distância Percorrida (m)")
plt.ylabel("Frequência Cardíaca (bpm)")

# Criar as linhas personalizadas para a legenda
custom_lines = [Line2D([0], [0], color='g', linestyle='--'),
                Line2D([0], [0], color='b', linestyle='--'),
                Line2D([0], [0], color='red', linewidth=2)]

# Adicionar legenda com valores máximo e médio
plt.legend(custom_lines, [f"Máx: {fc_max} bpm", f"Média: {fc_media:.1f} bpm", "Frequência Cardíaca"], loc="upper left")

#Linha media e max

plt.axhline(df["Freq_Cardiaca"].max(), color='g', linestyle='--')
plt.axhline(df["Freq_Cardiaca"].mean(), color='b', linestyle='--')


# Exibir o gráfico
plt.grid(True)
plt.show()
````

````python
# Função para determinar a zona de potência com base na potência (em watts)
def get_power_zone(power, zones):
    for zone, (lower, upper) in zones.items():
        if lower <= power <= upper:
            return zone
    return 'Fora de Zona'  # Caso a potência esteja fora das zonas

# Supondo que você tenha uma coluna 'Potência' no DataFrame
df['Zona_Potencia'] = df['Potência'].apply(lambda x: get_power_zone(x, zones_potencia))

# Calcular o tempo total e o tempo em cada zona de potência
total_tempo = df['Tempo'].max()  # Tempo total da atividade
tempo_por_zona = df['Zona_Potencia'].value_counts().sort_index()

# Calcular o percentual de tempo em cada zona
percentual_por_zona = (tempo_por_zona / len(df)) * 100

# Criar o gráfico
fig, ax = plt.subplots(figsize=(10, 6))

# Plotar a distribuição das zonas de potência
ax.bar(tempo_por_zona.index, percentual_por_zona, color='skyblue')

# Adicionar título e rótulos
ax.set_title('Distribuição das Zonas de Potência ao Longo da Atividade', fontsize=14)
ax.set_xlabel('Zona de Potência', fontsize=12)
ax.set_ylabel('Percentual de Tempo (%)', fontsize=12)

# Exibir os valores percentuais no topo de cada barra
for i, v in enumerate(percentual_por_zona):
    ax.text(i, v + 1, f'{v:.1f}%', ha='center', va='bottom', fontsize=10)

# Exibir o gráfico
plt.grid(True)
plt.show()
````
